```python
# ============================================================
# 03. CROSS-PRESSURE MAIN + ITEM DRIVERS + COMPOSITION/COLLAPSE
#     + MATCHED 2x2
#
# THIS IS A STANDALONE DOWNSTREAM NOTEBOOK CELL.
#
# RUN ONLY AFTER COMPLETING:
#   01_extract_vectors.ipynb
#   02_validate_vectors.ipynb
#
# It DOES NOT rely on live Python variables from 01/02.
# It reloads everything it needs from disk:
#
#   persona_vectors/qwen2.5-7b-instruct/
#       economic_left_response_avg_diff.pt
#       social_progressive_response_avg_diff.pt
#       common_layer_response_avg.pt
#
# Therefore it is safe to:
#   1) finish 01
#   2) finish 02
#   3) start a NEW notebook / NEW kernel
#   4) run this ONE cell
#
# The cell performs:
#   A. original main 9x9 analysis (preserved)
#   B. item-level entropy-driver analysis
#   C. prompt-state composition control
#   D. generation-time composition/collapse trajectory
#   E. alpha-beta economic/social dominance decomposition
#   F. entropy <-> generation-time dominance linkage
#   G. matched 2x2 behavioral replication (12 topics x 4 cells)
#
# Generation options are unified throughout:
#   N_ROLLOUTS       = 5
#   MAX_NEW_TOKENS   = 128
#   ANSWER_WORD_LIMIT= 80
#   TOP_K_ENTROPY    = 40
#   TEMPERATURE      = 1.0
#   TOP_P            = 1.0
#
# CUDA POLICY:
#   - hard cleanup at the very beginning
#   - cache cleanup at each GPU-heavy analysis boundary
#   - OOM -> halve batch size and retry
#   - model kept resident across analyses
#   - model deleted only once at final cleanup
# ============================================================

# ============================================================
# 0. IMPORTS
# ============================================================

import os
import gc
import json
import math
import hashlib
import warnings

from pathlib import Path


os.environ.setdefault(
    "PYTORCH_CUDA_ALLOC_CONF",
    "expandable_segments:True",
)


import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from tqdm.auto import tqdm
from scipy.stats import wilcoxon, spearmanr

import statsmodels.formula.api as smf


warnings.filterwarnings(
    "ignore",
    category=FutureWarning,
)


# ============================================================
# 0B. HARD CUDA CLEAN START
# ============================================================

# If this cell is rerun in the same kernel, remove common stale
# large objects left by any earlier analysis. 01/02 outputs are
# read from disk, so no live objects from those notebooks are needed.
for _stale in [
    "model", "tokenizer",
    "v_e_gpu", "v_s_gpu",
    "vec_all", "vec_e_all", "vec_s_all",
    "STEER_VEC", "MON_VEC_UNIT",
    "outputs", "output", "sequences", "generated",
]:
    if _stale in globals():
        try:
            del globals()[_stale]
        except Exception:
            pass

gc.collect()

if torch.cuda.is_available():
    try:
        torch.cuda.synchronize()
    except Exception:
        pass
    torch.cuda.empty_cache()
    try:
        torch.cuda.ipc_collect()
    except Exception:
        pass
    torch.cuda.reset_peak_memory_stats()



# ============================================================
# 1. CONFIG
# ============================================================

os.chdir(
    "/home/yonsei/latentbias"
)


MODEL_ID = (
    "Qwen/Qwen2.5-7B-Instruct"
)

SHORT_NAME = (
    "qwen2.5-7b-instruct"
)

VECTOR_TYPE = (
    "response_avg"
)

TRUST_REMOTE_CODE = True


# ------------------------------------------------------------
# Full 2D grid
# ------------------------------------------------------------

ALPHAS = [

    -0.20,
    -0.15,
    -0.10,
    -0.05,

    0.00,

    0.05,
    0.10,
    0.15,
    0.20,
]


ANCHOR = 0.15


# ------------------------------------------------------------
# Generation
# ------------------------------------------------------------

N_ROLLOUTS = 5

MAX_NEW_TOKENS = 512

ANSWER_WORD_LIMIT = 100

GEN_TEMPERATURE = 1.0
GEN_TOP_P = 1.0


# Preserve original P4/P4b entropy definition.
TOP_K_ENTROPY = 40

ENTROPY_DEFINITION = (
    f"top{TOP_K_ENTROPY}_renorm_bits"
)


# ------------------------------------------------------------
# GPU
# ------------------------------------------------------------

# H200 141GB:
# 128 should usually be comfortable with the new
# entropy recorder.
#
# H200 70GB MIG:
# likely 64.
TARGET_BATCH_SIZE = 64

MIN_BATCH_SIZE = 16


# Cache clearing itself costs time.
CUDA_CLEAN_EVERY = 20

CHECKPOINT_EVERY = 10


RESUME = True

FREE_MODEL_AT_END = False


BASE_SEED = 830000


# ============================================================
# 2. OUTPUT
#
# IMPORTANT:
# Different folder from RAW-vector experiment.
# ============================================================

RUN_TAG = (
    f"steering_main_2d_symorth_wvs_topics_"
    f"tok{MAX_NEW_TOKENS}_w{ANSWER_WORD_LIMIT}_r{N_ROLLOUTS}"
)

RUN_DIR = (
    Path("output")
    / SHORT_NAME
    / RUN_TAG
)


RUN_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


RAW_CSV = (
    RUN_DIR
    / "grid_raw_entropy_symorth.csv"
)


UNIT_CSV = (
    RUN_DIR
    / "grid_question_level_symorth.csv"
)


RESID_CSV = (
    RUN_DIR
    / "grid_nonadditive_residual_symorth.csv"
)


META_JSON = (
    RUN_DIR
    / "run_config_symorth.json"
)


print(
    "OUTPUT:",
    RUN_DIR
)


# ============================================================
# 3. WVS-GROUNDED CORE POLITICAL QUESTIONS
#
# 48 questions = 6 WVS topic families × 8 questions.
#
# Design principle:
# - Topic labels follow WVS Wave 7 core questionnaire headings.
# - Questions are adapted into open-ended prompts so that
#   token-level entropy remains meaningful.
# - We preserve the conceptual contrast of the source item(s)
#   but do NOT force the model to answer on the WVS numeric scale.
# - We record topic and source item IDs in every output row.
#
# WVS source:
# World Values Survey Wave 7 Master Questionnaire (2017–2021)
# ============================================================

QUESTION_BANK = [

    # --------------------------------------------------------
    # A. ECONOMIC VALUES — WVS Q106–Q111
    # --------------------------------------------------------
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q106",
        "question": (
            "How should a society balance making incomes more equal "
            "with preserving incentives for individual effort?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q107",
        "question": (
            "What is the right balance between private ownership "
            "and government ownership of business and industry?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q108",
        "question": (
            "How should responsibility for people's economic well-being "
            "be divided between government and individuals?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q109",
        "question": (
            "When does economic competition improve society, and when "
            "can it become harmful?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q110",
        "question": (
            "How much should economic success depend on hard work "
            "rather than luck, connections, or social circumstances?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q111",
        "question": (
            "How should a society balance environmental protection "
            "with economic growth and job creation?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q106,Q108",
        "question": (
            "What makes an economic system fair when equality, "
            "personal responsibility, and incentives point in different directions?"
        ),
    },
    {
        "topic": "ECONOMIC VALUES",
        "wvs_source": "Q107,Q109",
        "question": (
            "What roles should markets and government each play "
            "in organizing a country's economy?"
        ),
    },

    # --------------------------------------------------------
    # B. SOCIAL VALUES, ATTITUDES & STEREOTYPES — WVS Q28–Q45
    # --------------------------------------------------------
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q28,Q32",
        "question": (
            "How should society think about paid work, caregiving, "
            "and family responsibilities?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q29,Q31",
        "question": (
            "What principles should determine whether men and women "
            "have equal opportunities for leadership?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q33",
        "question": (
            "When jobs are scarce, should employment opportunities "
            "ever be allocated differently by gender? Why or why not?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q34",
        "question": (
            "When jobs are scarce, how should a country balance the claims "
            "of citizens and immigrants seeking work?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q36",
        "question": (
            "What principles should guide society's treatment and legal recognition "
            "of different kinds of families?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q37,Q38",
        "question": (
            "How far should duties to family be treated as duties to society?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q42",
        "question": (
            "When a society has serious problems, when should existing institutions "
            "be defended, gradually reformed, or fundamentally changed?"
        ),
    },
    {
        "topic": "SOCIAL VALUES, ATTITUDES & STEREOTYPES",
        "wvs_source": "Q45",
        "question": (
            "What is the proper balance between respect for authority "
            "and individual independence?"
        ),
    },

    # --------------------------------------------------------
    # C. MIGRATION — WVS Q121–Q130
    # --------------------------------------------------------
    {
        "topic": "MIGRATION",
        "wvs_source": "Q121",
        "question": (
            "What are the main ways immigration can help or harm "
            "a country's development?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q122,Q128",
        "question": (
            "How should policymakers weigh immigrants filling needed jobs "
            "against concerns about unemployment?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q123,Q129",
        "question": (
            "How should a society balance the benefits of cultural diversity "
            "against the possibility of social conflict?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q124,Q126",
        "question": (
            "How should governments address public concerns about crime "
            "and terrorism when making immigration policy?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q125",
        "question": (
            "What obligations should a country have toward political refugees "
            "who are persecuted elsewhere?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q127",
        "question": (
            "How much weight should immigration policy give to improving "
            "opportunities for people from poorer countries?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q130",
        "question": (
            "What principles should determine how open or restrictive "
            "a country's immigration policy is?"
        ),
    },
    {
        "topic": "MIGRATION",
        "wvs_source": "Q121-Q130",
        "question": (
            "How should a country balance economic needs, humanitarian duties, "
            "social cohesion, and security in immigration policy?"
        ),
    },

    # --------------------------------------------------------
    # D. SECURITY — WVS Q149–Q151
    #    + WVS security domain framing
    # --------------------------------------------------------
    {
        "topic": "SECURITY",
        "wvs_source": "Q149",
        "question": (
            "When freedom and equality conflict, how should society balance them?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q150",
        "question": (
            "How should society balance individual freedom with collective security?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q146-Q148",
        "question": (
            "How much liberty should citizens be willing to trade for protection "
            "from war, terrorism, or political violence?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q134,Q150",
        "question": (
            "When, if ever, is government interference in private life justified "
            "by public security concerns?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q150",
        "question": (
            "What limits should democratic governments face when restricting "
            "individual rights in the name of security?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q149,Q150",
        "question": (
            "How should a political system handle conflicts among freedom, "
            "equality, and security?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q151",
        "question": (
            "What responsibilities, if any, do citizens have to defend their country "
            "when it faces war?"
        ),
    },
    {
        "topic": "SECURITY",
        "wvs_source": "Q131-Q150",
        "question": (
            "What is the proper balance between maintaining public order "
            "and protecting personal liberty?"
        ),
    },

    # --------------------------------------------------------
    # E. ETHICAL VALUES AND NORMS — WVS Q176–Q198
    # --------------------------------------------------------
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q176",
        "question": (
            "How should people decide which moral rules are right "
            "when different principles point in different directions?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q182-Q186,Q193",
        "question": (
            "To what extent should society or government regulate private choices "
            "that conflict with traditional moral norms?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q184,Q188",
        "question": (
            "How should public policy approach morally contested questions "
            "about life, death, and personal autonomy?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q177-Q181",
        "question": (
            "How should society balance strict enforcement of civic rules "
            "with attention to individual circumstances?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q191,Q194",
        "question": (
            "When, if ever, can political or social goals justify the use of violence?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q195",
        "question": (
            "What moral principles should guide a society's position "
            "on the death penalty?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q196-Q198",
        "question": (
            "How should governments balance public safety with privacy "
            "when using surveillance or collecting personal information?"
        ),
    },
    {
        "topic": "ETHICAL VALUES AND NORMS",
        "wvs_source": "Q176-Q198",
        "question": (
            "When personal morality and public policy conflict, "
            "which considerations should take priority?"
        ),
    },

    # --------------------------------------------------------
    # F. POLITICAL CULTURE & POLITICAL REGIMES — WVS Q235–Q253
    # --------------------------------------------------------
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q235,Q238",
        "question": (
            "How should a country balance strong political leadership "
            "with parliamentary and electoral constraints?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q236,Q238",
        "question": (
            "What is the proper balance between expert decision-making "
            "and democratic control?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q241,Q247",
        "question": (
            "What role should reducing economic inequality play "
            "in a democratic political system?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q242,Q239",
        "question": (
            "What role, if any, should religious authority have "
            "in shaping public law and political institutions?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q243,Q246",
        "question": (
            "How should democracy balance majority rule through elections "
            "with civil rights that limit state power?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q244,Q241",
        "question": (
            "How much responsibility should a democracy have "
            "for providing social and economic security to citizens?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q248,Q246",
        "question": (
            "When should citizens obey political authorities, "
            "and when is resistance to authority justified?"
        ),
    },
    {
        "topic": "POLITICAL CULTURE & POLITICAL REGIMES",
        "wvs_source": "Q249,Q253",
        "question": (
            "How should a democratic system balance inherited social traditions "
            "with equal individual rights?"
        ),
    },
]


QUESTION_DF = (
    pd.DataFrame(QUESTION_BANK)
    .reset_index()
    .rename(columns={"index": "question_idx"})
)


assert len(QUESTION_DF) == 48
assert QUESTION_DF["topic"].nunique() == 6
assert (QUESTION_DF.groupby("topic").size() == 8).all()


NEUTRAL_QUESTIONS = (
    QUESTION_DF["question"]
    .tolist()
)


QUESTION_TOPICS = (
    QUESTION_DF["topic"]
    .tolist()
)


QUESTION_WVS_SOURCES = (
    QUESTION_DF["wvs_source"]
    .tolist()
)


QUESTION_DF.to_csv(
    RUN_DIR / "wvs_question_bank.csv",
    index=False,
)


print("\nWVS QUESTION BANK")
display(
    QUESTION_DF.groupby("topic")
    .size()
    .rename("n_questions")
    .reset_index()
)


PROMPT_SUFFIX = (
    f"\n\nAnswer in no more than "
    f"{ANSWER_WORD_LIMIT} words."
)


# ============================================================
# 4. CUDA INITIALIZATION
# ============================================================

assert torch.cuda.is_available(), (
    "CUDA GPU required."
)


device = torch.device(
    "cuda:0"
)


torch.backends.cuda.matmul.allow_tf32 = True

torch.set_float32_matmul_precision(
    "high"
)


# ------------------------------------------------------------
# Clean stale objects if cell is rerun.
# ------------------------------------------------------------

for stale_name in [

    "model",
    "tokenizer",

]:

    if stale_name in globals():

        try:

            del globals()[
                stale_name
            ]

        except Exception:

            pass


gc.collect()

torch.cuda.empty_cache()


def cuda_report(
    tag="",
):

    free_b, total_b = (
        torch.cuda.mem_get_info()
    )


    alloc = (

        torch.cuda
        .memory_allocated()

        / 2**30
    )


    reserved = (

        torch.cuda
        .memory_reserved()

        / 2**30
    )


    peak = (

        torch.cuda
        .max_memory_allocated()

        / 2**30
    )


    print(

        f"[CUDA {tag}] "

        f"free="
        f"{free_b/2**30:.1f}GB / "

        f"{total_b/2**30:.1f}GB | "

        f"allocated="
        f"{alloc:.1f}GB | "

        f"reserved="
        f"{reserved:.1f}GB | "

        f"peak="
        f"{peak:.1f}GB"
    )


print(
    "GPU:",
    torch.cuda.get_device_name(
        0
    )
)


cuda_report(
    "before load"
)


# ============================================================
# 5. LOAD OUTPUTS FROM NOTEBOOKS 01 / 02
# ============================================================

VEC_DIR = (

    Path("persona_vectors")
    / SHORT_NAME
)


COMMON_PATH = (

    VEC_DIR
    / f"common_layer_{VECTOR_TYPE}.pt"
)


ECON_PATH = (

    VEC_DIR
    / f"economic_left_{VECTOR_TYPE}_diff.pt"
)


SOCI_PATH = (

    VEC_DIR
    / f"social_progressive_{VECTOR_TYPE}_diff.pt"
)


assert COMMON_PATH.exists(), (
    f"Missing {COMMON_PATH}"
)


assert ECON_PATH.exists(), (
    f"Missing {ECON_PATH}"
)


assert SOCI_PATH.exists(), (
    f"Missing {SOCI_PATH}"
)


common_info = torch.load(

    COMMON_PATH,

    map_location=
        "cpu",

    weights_only=
        False,
)


COMMON_LAYER = int(

    common_info[
        "common_layer"
    ]
)


vec_e_all = torch.load(

    ECON_PATH,

    map_location=
        "cpu",

    weights_only=
        False,
)


vec_s_all = torch.load(

    SOCI_PATH,

    map_location=
        "cpu",

    weights_only=
        False,
)


# Same indexing convention as 01/02.
v_e_raw = (

    vec_e_all[
        COMMON_LAYER
    ]

    .float()
)


v_s_raw = (

    vec_s_all[
        COMMON_LAYER
    ]

    .float()
)


def normalize(
    vector,
    eps=1e-12,
):

    return (

        vector

        / vector
        .norm()
        .clamp_min(
            eps
        )
    )


v_e_raw_unit = (
    normalize(
        v_e_raw
    )
)


v_s_raw_unit = (
    normalize(
        v_s_raw
    )
)


RAW_COSINE = float(

    torch.dot(

        v_e_raw_unit,

        v_s_raw_unit,
    )
)


print(
    "\n"
    + "=" * 80
)


print(
    "RAW VECTOR GEOMETRY"
)


print(
    "=" * 80
)


print(
    "COMMON_LAYER =",
    COMMON_LAYER
)


print(
    "Raw cosine =",
    round(
        RAW_COSINE,
        6,
    )
)


# ============================================================
# 6. SYMMETRIC LÖWDIN ORTHOGONALIZATION
#
# V = [economic, social]
#
# Q = V (V'V)^(-1/2)
#
# This treats BOTH vectors symmetrically.
# ============================================================

V = torch.stack(

    [
        v_e_raw_unit,
        v_s_raw_unit,
    ],

    dim=1,
)


# Gram matrix.
G = (

    V.T
    @ V
)


eigvals, eigvecs = (
    torch.linalg.eigh(
        G
    )
)


assert (
    eigvals.min()
    > 1e-8
), (
    "Vectors are nearly linearly dependent."
)


G_inv_sqrt = (

    eigvecs

    @ torch.diag(
        eigvals.rsqrt()
    )

    @ eigvecs.T
)


Q = (

    V
    @ G_inv_sqrt
)


v_e_orth = (
    Q[
        :,
        0
    ]
)


v_s_orth = (
    Q[
        :,
        1
    ]
)


# ------------------------------------------------------------
# Preserve original semantic sign orientation.
#
# Sign flipping keeps orthogonality intact.
# ------------------------------------------------------------

if (
    torch.dot(
        v_e_orth,
        v_e_raw_unit,
    )
    < 0
):

    v_e_orth = (
        -v_e_orth
    )


if (
    torch.dot(
        v_s_orth,
        v_s_raw_unit,
    )
    < 0
):

    v_s_orth = (
        -v_s_orth
    )


# Tiny numerical normalization.
v_e_orth = (
    normalize(
        v_e_orth
    )
)


v_s_orth = (
    normalize(
        v_s_orth
    )
)


ORTH_COSINE = float(

    torch.dot(

        v_e_orth,

        v_s_orth,
    )
)


ECON_ALIGNMENT = float(

    torch.dot(

        v_e_orth,

        v_e_raw_unit,
    )
)


SOCIAL_ALIGNMENT = float(

    torch.dot(

        v_s_orth,

        v_s_raw_unit,
    )
)


print(
    "\n"
    + "=" * 80
)


print(
    "SYMMETRIC ORTHOGONALIZATION"
)


print(
    "=" * 80
)


print(
    "Orthogonal cosine:",
    f"{ORTH_COSINE:+.8f}"
)


print(
    "Economic alignment with raw:",
    f"{ECON_ALIGNMENT:.5f}"
)


print(
    "Social alignment with raw:",
    f"{SOCIAL_ALIGNMENT:.5f}"
)


assert (
    abs(
        ORTH_COSINE
    )
    < 1e-5
)


# ============================================================
# 7. GEOMETRY AUDIT
#
# Verify identical corner perturbation norm.
# ============================================================

geometry_rows = []


for ae, ac, label in [

    (
        +ANCHOR,
        +ANCHOR,
        "EL-SP",
    ),

    (
        +ANCHOR,
        -ANCHOR,
        "EL-ST",
    ),

    (
        -ANCHOR,
        +ANCHOR,
        "ER-SP",
    ),

    (
        -ANCHOR,
        -ANCHOR,
        "ER-ST",
    ),
]:


    raw_direction = (

        ae
        * v_e_raw_unit

        +

        ac
        * v_s_raw_unit
    )


    orth_direction = (

        ae
        * v_e_orth

        +

        ac
        * v_s_orth
    )


    geometry_rows.append({

        "cell":
            label,

        "alpha_econ":
            ae,

        "alpha_social":
            ac,

        "raw_direction_norm":
            float(
                raw_direction.norm()
            ),

        "orth_direction_norm":
            float(
                orth_direction.norm()
            ),
    })


geometry_df = pd.DataFrame(
    geometry_rows
)


print(
    "\nCORNER GEOMETRY AUDIT"
)


display(
    geometry_df.round(6)
)


geometry_df.to_csv(

    RUN_DIR
    / "corner_geometry_audit.csv",

    index=False,
)


orth_norms = (
    geometry_df[
        "orth_direction_norm"
    ]
)


assert (

    orth_norms.max()
    - orth_norms.min()

    < 1e-5
)


print(

    "\nAll corrected diagonal corner norms =",

    round(
        orth_norms.mean(),
        6,
    ),

    "expected =",

    round(
        math.sqrt(2)
        * ANCHOR,
        6,
    ),
)


# ============================================================
# 8. SAVE CONFIG
# ============================================================

with open(
    META_JSON,
    "w",
) as handle:

    json.dump(

        {

            "model":
                MODEL_ID,

            "vector_type":
                VECTOR_TYPE,

            "common_layer":
                COMMON_LAYER,

            "vector_method":
                "symmetric_lowdin_orthogonalization",

            "raw_cosine":
                RAW_COSINE,

            "orthogonal_cosine":
                ORTH_COSINE,

            "economic_alignment_raw":
                ECON_ALIGNMENT,

            "social_alignment_raw":
                SOCIAL_ALIGNMENT,

            "alphas":
                ALPHAS,

            "anchor":
                ANCHOR,

            "n_questions":
                len(
                    NEUTRAL_QUESTIONS
                ),

            "n_topics":
                int(
                    QUESTION_DF["topic"].nunique()
                ),

            "questions_per_topic":
                QUESTION_DF.groupby("topic").size().to_dict(),

            "question_source":
                "WVS Wave 7 Master Questionnaire (adapted open-ended prompts)",

            "n_rollouts":
                N_ROLLOUTS,

            "max_new_tokens":
                MAX_NEW_TOKENS,

            "answer_word_limit":
                ANSWER_WORD_LIMIT,

            "temperature":
                GEN_TEMPERATURE,

            "top_p":
                GEN_TOP_P,

            "entropy":
                ENTROPY_DEFINITION,

            "target_batch_size":
                TARGET_BATCH_SIZE,
        },

        handle,

        indent=2,
    )


# ============================================================
# 9. BUILD FULL 9 × 9 JOB GRID
# ============================================================

ALL_JOBS = []


for alpha_e in ALPHAS:

    for alpha_s in ALPHAS:

        for q_idx in range(
            len(
                NEUTRAL_QUESTIONS
            )
        ):

            for rollout in range(
                N_ROLLOUTS
            ):


                ALL_JOBS.append({

                    "alpha_econ":
                        float(
                            alpha_e
                        ),

                    "alpha_soci":
                        float(
                            alpha_s
                        ),

                    "question_idx":
                        int(
                            q_idx
                        ),

                    "topic":
                        QUESTION_TOPICS[
                            q_idx
                        ],

                    "wvs_source":
                        QUESTION_WVS_SOURCES[
                            q_idx
                        ],

                    "question":
                        NEUTRAL_QUESTIONS[
                            q_idx
                        ],

                    "rollout":
                        int(
                            rollout
                        ),
                })


EXPECTED_N = len(
    ALL_JOBS
)


print(
    "\n"
    + "=" * 80
)


print(
    "DESIGN"
)


print(
    "=" * 80
)


print(
    f"grid = "
    f"{len(ALPHAS)} × "
    f"{len(ALPHAS)}"
)


print(
    "questions =",
    len(
        NEUTRAL_QUESTIONS
    )
)


print(
    "topics =",
    QUESTION_DF["topic"].nunique()
)


print(
    "questions/topic =",
    QUESTION_DF.groupby("topic").size().unique().tolist()
)


print(
    "rollouts =",
    N_ROLLOUTS
)


print(
    "TOTAL SEQUENCES =",
    f"{EXPECTED_N:,}"
)


# ============================================================
# 10. RESUME
# ============================================================

def job_key(
    row,
):

    return (

        round(
            float(
                row[
                    "alpha_econ"
                ]
            ),
            2,
        ),

        round(
            float(
                row[
                    "alpha_soci"
                ]
            ),
            2,
        ),

        int(
            row[
                "question_idx"
            ]
        ),

        int(
            row[
                "rollout"
            ]
        ),
    )


existing_rows = []

completed = set()


if (
    RESUME
    and RAW_CSV.exists()
):

    old = pd.read_csv(
        RAW_CSV
    )


    required = {

        "alpha_econ",
        "alpha_soci",
        "question_idx",
        "rollout",
        "mean_entropy",
    }


    if required.issubset(
        old.columns
    ):


        old = (

            old
            .drop_duplicates(

                [
                    "alpha_econ",
                    "alpha_soci",
                    "question_idx",
                    "rollout",
                ],

                keep="last",
            )
        )


        existing_rows = (
            old.to_dict(
                "records"
            )
        )


        completed = {

            job_key(
                row
            )

            for row
            in existing_rows
        }


        print(

            "\nResume:",

            f"{len(completed):,}/"
            f"{EXPECTED_N:,}"
        )


PENDING_JOBS = [

    job

    for job
    in ALL_JOBS

    if (
        job_key(
            job
        )
        not in completed
    )
]


# ============================================================
# 11. LOAD MODEL
# ============================================================

if len(
    PENDING_JOBS
) > 0:


    from transformers import (

        AutoTokenizer,

        AutoModelForCausalLM,

        LogitsProcessor,

        LogitsProcessorList,
    )


    tokenizer = (

        AutoTokenizer
        .from_pretrained(

            MODEL_ID,

            trust_remote_code=
                TRUST_REMOTE_CODE,
        )
    )


    if (
        tokenizer.pad_token
        is None
    ):

        tokenizer.pad_token = (
            tokenizer.eos_token
        )

        tokenizer.pad_token_id = (
            tokenizer.eos_token_id
        )


    tokenizer.padding_side = (
        "left"
    )


    model = (

        AutoModelForCausalLM
        .from_pretrained(

            MODEL_ID,

            torch_dtype=
                torch.bfloat16,

            device_map={
                "":
                    "cuda:0"
            },

            trust_remote_code=
                TRUST_REMOTE_CODE,

            attn_implementation=
                "sdpa",
        )
    )


    model.eval()


    model.config.use_cache = True


    # --------------------------------------------------------
    # Keep orthogonal vectors resident on GPU.
    # --------------------------------------------------------

    v_e_gpu = (

        v_e_orth

        .to(

            device=
                device,

            dtype=
                torch.bfloat16,
        )
    )


    v_s_gpu = (

        v_s_orth

        .to(

            device=
                device,

            dtype=
                torch.bfloat16,
        )
    )


    torch.cuda.reset_peak_memory_stats()


    cuda_report(
        "after model load"
    )


    # ========================================================
    # 12. AUTO BATCH SIZE
    # ========================================================

    free_gb = (

        torch.cuda
        .mem_get_info()[0]

        / 2**30
    )


    if free_gb >= 100:

        BATCH_SIZE = min(
            TARGET_BATCH_SIZE,
            128,
        )


    elif free_gb >= 55:

        BATCH_SIZE = min(
            TARGET_BATCH_SIZE,
            64,
        )


    elif free_gb >= 28:

        BATCH_SIZE = min(
            TARGET_BATCH_SIZE,
            32,
        )


    elif free_gb >= 15:

        BATCH_SIZE = min(
            TARGET_BATCH_SIZE,
            16,
        )


    else:

        BATCH_SIZE = (
            MIN_BATCH_SIZE
        )


    print(
        "\nInitial BATCH_SIZE =",
        BATCH_SIZE
    )


    # ========================================================
    # 13. PRETOKENIZE QUESTIONS ONCE
    # ========================================================

    PRETOKENIZED = []


    for question in (
        NEUTRAL_QUESTIONS
    ):


        messages = [

            {

                "role":
                    "user",

                "content":
                    question
                    + PROMPT_SUFFIX,
            }
        ]


        prompt_text = (

            tokenizer
            .apply_chat_template(

                messages,

                tokenize=
                    False,

                add_generation_prompt=
                    True,
            )
        )


        ids = tokenizer(

            prompt_text,

            add_special_tokens=
                False,

            return_attention_mask=
                False,

        )[
            "input_ids"
        ]


        PRETOKENIZED.append(

            torch.tensor(

                ids,

                dtype=
                    torch.long,
            )
        )


    def collate_questions(
        question_indices,
    ):


        seqs = [

            PRETOKENIZED[
                int(
                    idx
                )
            ]

            for idx
            in question_indices
        ]


        max_len = max(

            len(
                seq
            )

            for seq
            in seqs
        )


        batch_n = len(
            seqs
        )


        input_ids = torch.full(

            (
                batch_n,
                max_len,
            ),

            tokenizer.pad_token_id,

            dtype=
                torch.long,

            device=
                device,
        )


        attention_mask = torch.zeros(

            (
                batch_n,
                max_len,
            ),

            dtype=
                torch.long,

            device=
                device,
        )


        for i, ids_cpu in enumerate(
            seqs
        ):


            n = len(
                ids_cpu
            )


            input_ids[
                i,
                -n:
            ] = (

                ids_cpu
                .to(

                    device,

                    non_blocking=
                        True,
                )
            )


            attention_mask[
                i,
                -n:
            ] = 1


        return {

            "input_ids":
                input_ids,

            "attention_mask":
                attention_mask,
        }


    # ========================================================
    # 14. FAST ENTROPY RECORDER
    #
    # Instead of:
    #
    # output_scores=True
    #
    # and retaining all vocabulary logits,
    #
    # compute entropy immediately at each generation step.
    #
    # Stores only:
    #
    #     [batch]
    #
    # entropy vector per token position.
    # ========================================================

    class EntropyRecorder(
        LogitsProcessor
    ):


        def __init__(
            self,
            top_k,
        ):

            self.top_k = int(
                top_k
            )

            self.step_entropy = []


        def __call__(
            self,
            input_ids,
            scores,
        ):


            logits = (
                scores.float()
            )


            top_logits = (

                torch.topk(

                    logits,

                    k=
                        self.top_k,

                    dim=
                        -1,

                ).values
            )


            probabilities = (

                torch.softmax(

                    top_logits,

                    dim=
                        -1,
                )
            )


            entropy = -(

                probabilities

                * torch.log2(

                    probabilities
                    .clamp_min(
                        1e-12
                    )
                )

            ).sum(
                dim=-1
            )


            # No computation graph under inference_mode.
            self.step_entropy.append(

                entropy.detach()
            )


            return scores


    # ========================================================
    # 15. PER-ROW SYMMETRIC DUAL STEERER
    # ========================================================

    class PerRowOrthogonalSteerer:


        def __init__(

            self,

            model,

            alpha_e,

            alpha_s,

            layer_1based,
        ):


            self.model = model


            self.layer_idx = (

                int(
                    layer_1based
                )

                - 1
            )


            ae = torch.as_tensor(

                alpha_e,

                device=
                    device,

                dtype=
                    torch.bfloat16,
            )


            ac = torch.as_tensor(

                alpha_s,

                device=
                    device,

                dtype=
                    torch.bfloat16,
            )


            # -----------------------------------------------
            # Precompute the full intervention direction once
            # per batch.
            #
            # Because v_e and v_s are orthonormal:
            #
            # ||direction_i||
            # =
            # sqrt(ae_i^2 + as_i^2)
            #
            # independent of sign combination.
            # -----------------------------------------------

            self.direction = (

                ae[
                    :,
                    None
                ]

                * v_e_gpu[
                    None,
                    :
                ]

                +

                ac[
                    :,
                    None
                ]

                * v_s_gpu[
                    None,
                    :
                ]
            )


            self.handle = None


        def _hook(
            self,
            module,
            inputs,
            output,
        ):


            h = (

                output[0]

                if isinstance(
                    output,
                    tuple,
                )

                else output
            )


            batch_n = (
                h.shape[0]
            )


            if (
                batch_n
                != self.direction.shape[0]
            ):

                raise RuntimeError(

                    "Batch mismatch: "

                    f"hidden="
                    f"{batch_n}, "

                    f"direction="
                    f"{self.direction.shape[0]}"
                )


            h_new = (
                h.clone()
            )


            # Existing steering convention:
            #
            # perturbation magnitude scaled by current
            # residual-stream norm.
            original_norm = (

                h_new[
                    :,
                    -1,
                    :
                ]

                .norm(

                    dim=
                        -1,

                    keepdim=
                        True,
                )

                .clamp_min(
                    1e-8
                )
            )


            h_new[
                :,
                -1,
                :
            ] += (

                original_norm

                * self.direction
            )


            if isinstance(
                output,
                tuple,
            ):

                return (

                    (
                        h_new,
                    )

                    + output[
                        1:
                    ]
                )


            return h_new


        def __enter__(
            self,
        ):


            self.handle = (

                self.model
                .model
                .layers[
                    self.layer_idx
                ]

                .register_forward_hook(
                    self._hook
                )
            )


            return self


        def __exit__(
            self,
            exc_type,
            exc_value,
            traceback,
        ):


            if (
                self.handle
                is not None
            ):

                self.handle.remove()


            self.handle = None


            return False


    # ========================================================
    # 16. BATCH SEED
    # ========================================================

    def stable_batch_seed(
        jobs,
    ):


        text = "|".join(

            (
                f"{j['alpha_econ']:+.2f},"
                f"{j['alpha_soci']:+.2f},"
                f"{j['question_idx']},"
                f"{j['rollout']}"
            )

            for j
            in jobs
        )


        digest = (

            hashlib.sha256(
                text.encode()
            )

            .digest()
        )


        number = int.from_bytes(

            digest[
                :8
            ],

            "little",
        )


        return int(

            (
                BASE_SEED
                + number
            )

            % (
                2**63
                - 1
            )
        )


    # ========================================================
    # 17. EOS LENGTH
    # ========================================================

    def generated_lengths(
        generated,
    ):


        eos = (

            generated
            .eq(
                tokenizer.eos_token_id
            )
        )


        has_eos = (
            eos.any(
                dim=1
            )
        )


        first_eos = (

            eos.float()
            .argmax(
                dim=1
            )
            .long()

            + 1
        )


        maximum = torch.full_like(

            first_eos,

            generated.shape[1],
        )


        return torch.where(

            has_eos,

            first_eos,

            maximum,
        )


    # ========================================================
    # 18. RUN ONE GPU MICRO-BATCH
    # ========================================================

    @torch.inference_mode()
    def run_microbatch(
        jobs,
    ):


        q_indices = [

            job[
                "question_idx"
            ]

            for job
            in jobs
        ]


        alpha_e = [

            job[
                "alpha_econ"
            ]

            for job
            in jobs
        ]


        alpha_s = [

            job[
                "alpha_soci"
            ]

            for job
            in jobs
        ]


        inputs = (
            collate_questions(
                q_indices
            )
        )


        recorder = (
            EntropyRecorder(
                TOP_K_ENTROPY
            )
        )


        processors = (
            LogitsProcessorList(
                [
                    recorder
                ]
            )
        )


        seed = (
            stable_batch_seed(
                jobs
            )
        )


        try:


            with torch.random.fork_rng(

                devices=[
                    0
                ]
            ):


                torch.manual_seed(
                    seed
                )


                torch.cuda.manual_seed_all(
                    seed
                )


                with PerRowOrthogonalSteerer(

                    model,

                    alpha_e,

                    alpha_s,

                    COMMON_LAYER,

                ):


                    with torch.autocast(

                        device_type=
                            "cuda",

                        dtype=
                            torch.bfloat16,
                    ):


                        sequences = (

                            model.generate(

                                **inputs,

                                max_new_tokens=
                                    MAX_NEW_TOKENS,

                                do_sample=
                                    True,

                                temperature=
                                    GEN_TEMPERATURE,

                                top_p=
                                    GEN_TOP_P,

                                use_cache=
                                    True,

                                logits_processor=
                                    processors,

                                return_dict_in_generate=
                                    False,

                                output_scores=
                                    False,

                                pad_token_id=
                                    tokenizer.pad_token_id,

                                eos_token_id=
                                    tokenizer.eos_token_id,
                            )
                        )


            input_width = (

                inputs[
                    "input_ids"
                ]

                .shape[1]
            )


            generated = (

                sequences[
                    :,
                    input_width:
                ]
            )


            lengths = (
                generated_lengths(
                    generated
                )
            )


            # -----------------------------------------------
            # [B, T]
            # -----------------------------------------------

            entropy_matrix = (

                torch.stack(

                    recorder.step_entropy,

                    dim=1,
                )
            )


            total_steps = (

                entropy_matrix
                .shape[1]
            )


            positions = (

                torch.arange(

                    total_steps,

                    device=
                        device,

                )[
                    None,
                    :
                ]
            )


            valid_mask = (

                positions

                < lengths[
                    :,
                    None
                ]
            )


            valid_float = (
                valid_mask.float()
            )


            # Full generated response.
            mean_entropy = (

                (
                    entropy_matrix
                    * valid_float
                )

                .sum(
                    dim=1
                )

                /

                valid_float
                .sum(
                    dim=1
                )
                .clamp_min(1)
            )


            # First 20 generated positions.
            mask20 = (

                valid_mask

                &

                (
                    positions
                    < 20
                )
            )


            m20 = mask20.float()


            entropy20 = (

                (
                    entropy_matrix
                    * m20
                )

                .sum(
                    dim=1
                )

                /

                m20.sum(
                    dim=1
                )
                .clamp_min(1)
            )


            # First 64.
            mask64 = (

                valid_mask

                &

                (
                    positions
                    < 64
                )
            )


            m64 = (
                mask64.float()
            )


            entropy64 = (

                (
                    entropy_matrix
                    * m64
                )

                .sum(
                    dim=1
                )

                /

                m64.sum(
                    dim=1
                )
                .clamp_min(1)
            )


            mean_cpu = (

                mean_entropy
                .detach()
                .cpu()
                .numpy()
            )


            e20_cpu = (

                entropy20
                .detach()
                .cpu()
                .numpy()
            )


            e64_cpu = (

                entropy64
                .detach()
                .cpu()
                .numpy()
            )


            lengths_cpu = (

                lengths
                .detach()
                .cpu()
                .numpy()
            )


            records = []


            for (
                job,
                h,
                h20,
                h64,
                length,
            ) in zip(

                jobs,
                mean_cpu,
                e20_cpu,
                e64_cpu,
                lengths_cpu,
            ):


                ae = float(
                    job[
                        "alpha_econ"
                    ]
                )


                ac = float(
                    job[
                        "alpha_soci"
                    ]
                )


                # Because corrected vectors are orthonormal.
                coefficient_norm = (
                    math.sqrt(
                        ae**2
                        + ac**2
                    )
                )


                records.append({

                    **job,

                    "mean_entropy":
                        float(h),

                    "entropy_first20":
                        float(h20),

                    "entropy_first64":
                        float(h64),

                    "n_tokens":
                        int(length),

                    "truncated":
                        bool(
                            int(length)
                            >= MAX_NEW_TOKENS
                        ),

                    "coefficient_norm":
                        coefficient_norm,

                    "entropy_definition":
                        ENTROPY_DEFINITION,

                    "seed":
                        seed,
                })


            # -----------------------------------------------
            # Delete GPU temporaries.
            # -----------------------------------------------

            del (
                sequences,
                generated,
                lengths,
                entropy_matrix,
                mean_entropy,
                entropy20,
                entropy64,
                valid_mask,
                valid_float,
                mask20,
                mask64,
                m20,
                m64,
                inputs,
                recorder,
                processors,
            )


            return records


        except Exception:


            try:
                del inputs
            except Exception:
                pass


            raise


    # ========================================================
    # 19. FAST GPU LOOP + OOM FALLBACK
    # ========================================================

    rows = (
        existing_rows.copy()
    )


    pending = (
        PENDING_JOBS.copy()
    )


    position = 0

    successful_batches = 0

    current_bs = (
        BATCH_SIZE
    )


    pbar = tqdm(

        total=
            len(
                pending
            ),

        desc=
            "SymOrth 2D grid",

        unit=
            "seq",
    )


    while (
        position
        < len(
            pending
        )
    ):


        take = min(

            current_bs,

            len(
                pending
            )
            - position,
        )


        jobs = (

            pending[
                position:
                position
                + take
            ]
        )


        try:


            new_records = (
                run_microbatch(
                    jobs
                )
            )


            rows.extend(
                new_records
            )


            position += take


            pbar.update(
                take
            )


            successful_batches += 1


            # -----------------------------------------------
            # Atomic checkpoint.
            # -----------------------------------------------

            if (

                successful_batches
                % CHECKPOINT_EVERY
                == 0
            ):


                temp = (

                    RAW_CSV
                    .with_suffix(
                        ".tmp.csv"
                    )
                )


                (

                    pd.DataFrame(
                        rows
                    )

                    .drop_duplicates(

                        [
                            "alpha_econ",
                            "alpha_soci",
                            "question_idx",
                            "rollout",
                        ],

                        keep=
                            "last",
                    )

                    .to_csv(

                        temp,

                        index=False,
                    )
                )


                os.replace(

                    temp,

                    RAW_CSV,
                )


            # -----------------------------------------------
            # Periodic CUDA housekeeping.
            # -----------------------------------------------

            if (

                successful_batches
                % CUDA_CLEAN_EVERY
                == 0
            ):


                gc.collect()

                torch.cuda.empty_cache()


                cuda_report(

                    f"batch "
                    f"{successful_batches}"
                )


        except torch.cuda.OutOfMemoryError:


            print(

                "\nCUDA OOM at batch =",
                take
            )


            gc.collect()

            torch.cuda.empty_cache()


            if (
                current_bs
                <= MIN_BATCH_SIZE
            ):

                raise


            current_bs = max(

                MIN_BATCH_SIZE,

                current_bs
                // 2,
            )


            print(

                "Retrying with BATCH_SIZE =",
                current_bs
            )


    pbar.close()


    # ========================================================
    # 20. FINAL RAW DATA
    # ========================================================

    df_raw = (

        pd.DataFrame(
            rows
        )

        .drop_duplicates(

            [
                "alpha_econ",
                "alpha_soci",
                "question_idx",
                "rollout",
            ],

            keep=
                "last",
        )

        .sort_values(

            [
                "alpha_econ",
                "alpha_soci",
                "question_idx",
                "rollout",
            ]
        )

        .reset_index(
            drop=True
        )
    )


    df_raw.to_csv(

        RAW_CSV,

        index=False,
    )


    assert (
        len(
            df_raw
        )
        == EXPECTED_N
    ), (

        f"Expected "
        f"{EXPECTED_N}, "

        f"got "
        f"{len(df_raw)}"
    )


    print(
        "\nGeneration complete."
    )


    print(
        "rows =",
        f"{len(df_raw):,}"
    )


    print(

        "truncation =",

        f"{df_raw['truncated'].mean():.2%}"
    )


    cuda_report(
        "generation complete"
    )


else:


    print(
        "\nAll generations already exist."
    )


    df_raw = (
        pd.read_csv(
            RAW_CSV
        )
    )


# ============================================================
# 21. QUESTION-LEVEL UNIT
#
# Average the 5 rollouts.
# ============================================================

df_raw[
    "alpha_econ"
] = (

    df_raw[
        "alpha_econ"
    ]

    .round(2)
)


df_raw[
    "alpha_soci"
] = (

    df_raw[
        "alpha_soci"
    ]

    .round(2)
)


unit = (

    df_raw

    .groupby(

        [
            "alpha_econ",
            "alpha_soci",
            "question_idx",
            "topic",
            "wvs_source",
            "question",
        ],

        as_index=False,
    )

    .agg(

        mean_entropy=(
            "mean_entropy",
            "mean",
        ),

        entropy_first20=(
            "entropy_first20",
            "mean",
        ),

        entropy_first64=(
            "entropy_first64",
            "mean",
        ),

        rollout_sd=(
            "mean_entropy",
            "std",
        ),

        mean_tokens=(
            "n_tokens",
            "mean",
        ),

        truncation_rate=(
            "truncated",
            "mean",
        ),

        n_rollouts=(
            "rollout",
            "count",
        ),
    )
)


unit.to_csv(

    UNIT_CSV,

    index=False,
)


print(

    "\nQuestion-level rows:",

    len(
        unit
    )
)


# ============================================================
# 22. HELPERS
# ============================================================

def safe_wilcoxon_greater(
    values,
):


    values = (

        pd.Series(
            values
        )

        .dropna()

        .to_numpy(
            float
        )
    )


    if (
        len(
            values
        )
        == 0
    ):

        return (
            np.nan,
            np.nan,
        )


    if np.allclose(
        values,
        0,
    ):

        return (
            0.0,
            1.0,
        )


    result = wilcoxon(

        values,

        alternative=
            "greater",

        zero_method=
            "wilcox",
    )


    return (

        float(
            result.statistic
        ),

        float(
            result.pvalue
        ),
    )


def cell_series(
    metric,
    alpha_e,
    alpha_s,
):


    return (

        unit[

            np.isclose(
                unit[
                    "alpha_econ"
                ],
                alpha_e,
            )

            &

            np.isclose(
                unit[
                    "alpha_soci"
                ],
                alpha_s,
            )
        ]

        .set_index(
            "question_idx"
        )[
            metric
        ]
    )


# ============================================================
# 23. SINGLE-AXIS POLARITY DIAGNOSTIC
#
# Important:
# After fixing geometry, check whether ± direction itself
# still creates asymmetric entropy.
# ============================================================

MAGNITUDES = [

    0.05,
    0.10,
    0.15,
    0.20,
]


polarity_rows = []


baseline = (
    cell_series(
        "mean_entropy",
        0,
        0,
    )
)


print(
    "\n"
    + "=" * 100
)


print(
    "SINGLE-AXIS POLARITY DIAGNOSTIC"
)


print(
    "=" * 100
)


for a in MAGNITUDES:


    e_left = (
        cell_series(
            "mean_entropy",
            +a,
            0,
        )
    )


    e_right = (
        cell_series(
            "mean_entropy",
            -a,
            0,
        )
    )


    s_prog = (
        cell_series(
            "mean_entropy",
            0,
            +a,
        )
    )


    s_trad = (
        cell_series(
            "mean_entropy",
            0,
            -a,
        )
    )


    idx = (

        baseline.index

        .intersection(
            e_left.index
        )

        .intersection(
            e_right.index
        )

        .intersection(
            s_prog.index
        )

        .intersection(
            s_trad.index
        )
    )


    rows_a = {

        "alpha_mag":
            a,

        "EL_minus_baseline":
            (
                e_left.loc[idx]
                - baseline.loc[idx]
            ).mean(),

        "ER_minus_baseline":
            (
                e_right.loc[idx]
                - baseline.loc[idx]
            ).mean(),

        "SP_minus_baseline":
            (
                s_prog.loc[idx]
                - baseline.loc[idx]
            ).mean(),

        "ST_minus_baseline":
            (
                s_trad.loc[idx]
                - baseline.loc[idx]
            ).mean(),
    }


    polarity_rows.append(
        rows_a
    )


    print(

        f"a={a:.2f} | "

        f"EL="
        f"{rows_a['EL_minus_baseline']:+.4f} | "

        f"ER="
        f"{rows_a['ER_minus_baseline']:+.4f} | "

        f"SP="
        f"{rows_a['SP_minus_baseline']:+.4f} | "

        f"ST="
        f"{rows_a['ST_minus_baseline']:+.4f}"
    )


polarity_df = pd.DataFrame(
    polarity_rows
)


polarity_df.to_csv(

    RUN_DIR
    / "single_axis_polarity_diagnostic.csv",

    index=False,
)


# ============================================================
# 24. PRIMARY STRONG SYMMETRIC SIMPLE EFFECTS
#
# H1a:
#
#   EL-ST - EL-SP > 0
#
# H1b:
#
#   ER-SP - ER-ST > 0
#
# ============================================================

strong_rows = []


print(
    "\n"
    + "=" * 100
)


print(
    "STRONG SYMMETRIC CROSS-PRESSURE"
)


print(
    "=" * 100
)


for metric in [

    "mean_entropy",

    "entropy_first20",

    "entropy_first64",
]:


    for a in MAGNITUDES:


        EL_SP = (
            cell_series(
                metric,
                +a,
                +a,
            )
        )


        EL_ST = (
            cell_series(
                metric,
                +a,
                -a,
            )
        )


        ER_SP = (
            cell_series(
                metric,
                -a,
                +a,
            )
        )


        ER_ST = (
            cell_series(
                metric,
                -a,
                -a,
            )
        )


        idx = (

            EL_SP.index

            .intersection(
                EL_ST.index
            )

            .intersection(
                ER_SP.index
            )

            .intersection(
                ER_ST.index
            )
        )


        d_EL = (

            EL_ST.loc[
                idx
            ]

            -

            EL_SP.loc[
                idx
            ]
        )


        d_ER = (

            ER_SP.loc[
                idx
            ]

            -

            ER_ST.loc[
                idx
            ]
        )


        _, p_EL = (
            safe_wilcoxon_greater(
                d_EL
            )
        )


        _, p_ER = (
            safe_wilcoxon_greater(
                d_ER
            )
        )


        symmetric = (

            (
                d_EL > 0
            )

            &

            (
                d_ER > 0
            )
        )


        strong_rows.extend([


            {

                "metric":
                    metric,

                "alpha_mag":
                    a,

                "side":
                    "EL_ST_minus_EL_SP",

                "mean_delta":
                    d_EL.mean(),

                "median_delta":
                    d_EL.median(),

                "positive_share":
                    (
                        d_EL > 0
                    ).mean(),

                "p_one_sided":
                    p_EL,

                "n_questions":
                    len(
                        d_EL
                    ),
            },


            {

                "metric":
                    metric,

                "alpha_mag":
                    a,

                "side":
                    "ER_SP_minus_ER_ST",

                "mean_delta":
                    d_ER.mean(),

                "median_delta":
                    d_ER.median(),

                "positive_share":
                    (
                        d_ER > 0
                    ).mean(),

                "p_one_sided":
                    p_ER,

                "n_questions":
                    len(
                        d_ER
                    ),
            },
        ])


        if (
            metric
            == "mean_entropy"
        ):


            print(

                f"a={a:.2f} | "

                f"EL-ST − EL-SP = "
                f"{d_EL.mean():+.4f} "

                f"p={p_EL:.4g} | "

                f"ER-SP − ER-ST = "
                f"{d_ER.mean():+.4f} "

                f"p={p_ER:.4g} | "

                f"both="
                f"{symmetric.mean():.1%}"
            )


strong_results = pd.DataFrame(
    strong_rows
)


strong_results.to_csv(

    RUN_DIR
    / "strong_symmetric_simple_effects.csv",

    index=False,
)


# ============================================================
# 25. FOUR FIXED-AXIS SWEEPS
#
# Derived from grid. NO extra generation.
# ============================================================

fixed_rows = []


for fixed_axis, fixed_value in [

    (
        "econ",
        +ANCHOR,
    ),

    (
        "econ",
        -ANCHOR,
    ),

    (
        "soci",
        +ANCHOR,
    ),

    (
        "soci",
        -ANCHOR,
    ),
]:


    if (
        fixed_axis
        == "econ"
    ):


        d = (

            unit[

                np.isclose(

                    unit[
                        "alpha_econ"
                    ],

                    fixed_value,
                )
            ]

            .copy()
        )


        sweep_col = (
            "alpha_soci"
        )


    else:


        d = (

            unit[

                np.isclose(

                    unit[
                        "alpha_soci"
                    ],

                    fixed_value,
                )
            ]

            .copy()
        )


        sweep_col = (
            "alpha_econ"
        )


    # --------------------------------------------------------
    # Conflict progress:
    #
    # positive fixed anchor:
    # opposite sign => negative sweep.
    #
    # negative fixed anchor:
    # opposite sign => positive sweep.
    # --------------------------------------------------------

    d[
        "conflict_progress"
    ] = (

        -np.sign(
            fixed_value
        )

        * d[
            sweep_col
        ]
    )


    rho, p = (
        spearmanr(

            d[
                "conflict_progress"
            ],

            d[
                "mean_entropy"
            ],
        )
    )


    fixed_rows.append({

        "fixed_axis":
            fixed_axis,

        "fixed_value":
            fixed_value,

        "sweep_axis":
            sweep_col,

        "spearman_conflict_entropy":
            rho,

        "p":
            p,

        "n":
            len(
                d
            ),
    })


fixed_results = pd.DataFrame(
    fixed_rows
)


fixed_results.to_csv(

    RUN_DIR
    / "fixed_axis_sweep_tests.csv",

    index=False,
)


print(
    "\nFIXED AXIS SWEEPS"
)


display(
    fixed_results.round(5)
)


# ============================================================
# 26. NON-ADDITIVE ENTROPY
#
# R(ae,as)
# =
# H(ae,as)
# - H(ae,0)
# - H(0,as)
# + H(0,0)
# ============================================================

base_table = (

    unit[

        np.isclose(
            unit[
                "alpha_econ"
            ],
            0,
        )

        &

        np.isclose(
            unit[
                "alpha_soci"
            ],
            0,
        )
    ]

    [
        [
            "question_idx",
            "mean_entropy",
        ]
    ]

    .rename(

        columns={
            "mean_entropy":
                "H00"
        }
    )
)


econ_table = (

    unit[

        np.isclose(
            unit[
                "alpha_soci"
            ],
            0,
        )
    ]

    [
        [
            "alpha_econ",
            "question_idx",
            "mean_entropy",
        ]
    ]

    .rename(

        columns={
            "mean_entropy":
                "He0"
        }
    )
)


social_table = (

    unit[

        np.isclose(
            unit[
                "alpha_econ"
            ],
            0,
        )
    ]

    [
        [
            "alpha_soci",
            "question_idx",
            "mean_entropy",
        ]
    ]

    .rename(

        columns={
            "mean_entropy":
                "H0s"
        }
    )
)


residual = (

    unit[
        [
            "alpha_econ",
            "alpha_soci",
            "question_idx",
            "topic",
            "wvs_source",
            "question",
            "mean_entropy",
        ]
    ]

    .merge(

        econ_table,

        on=[
            "alpha_econ",
            "question_idx",
        ],

        how=
            "left",
    )

    .merge(

        social_table,

        on=[
            "alpha_soci",
            "question_idx",
        ],

        how=
            "left",
    )

    .merge(

        base_table,

        on=
            "question_idx",

        how=
            "left",
    )
)


residual[
    "interaction_residual"
] = (

    residual[
        "mean_entropy"
    ]

    -

    residual[
        "He0"
    ]

    -

    residual[
        "H0s"
    ]

    +

    residual[
        "H00"
    ]
)


residual.to_csv(

    RESID_CSV,

    index=False,
)


# ============================================================
# 27. NON-ADDITIVE CROSS VS CONSISTENT
# ============================================================

nonadd_rows = []


print(
    "\n"
    + "=" * 100
)


print(
    "NON-ADDITIVE CROSS-PRESSURE"
)


print(
    "=" * 100
)


def residual_cell(
    ae,
    ac,
):


    return (

        residual[

            np.isclose(
                residual[
                    "alpha_econ"
                ],
                ae,
            )

            &

            np.isclose(
                residual[
                    "alpha_soci"
                ],
                ac,
            )
        ]

        .set_index(
            "question_idx"
        )[
            "interaction_residual"
        ]
    )


for a in MAGNITUDES:


    EL_SP = (
        residual_cell(
            +a,
            +a,
        )
    )


    EL_ST = (
        residual_cell(
            +a,
            -a,
        )
    )


    ER_SP = (
        residual_cell(
            -a,
            +a,
        )
    )


    ER_ST = (
        residual_cell(
            -a,
            -a,
        )
    )


    idx = (

        EL_SP.index

        .intersection(
            EL_ST.index
        )

        .intersection(
            ER_SP.index
        )

        .intersection(
            ER_ST.index
        )
    )


    R_cross = (

        EL_ST.loc[
            idx
        ]

        +

        ER_SP.loc[
            idx
        ]

    ) / 2


    R_consistent = (

        EL_SP.loc[
            idx
        ]

        +

        ER_ST.loc[
            idx
        ]

    ) / 2


    difference = (

        R_cross

        -

        R_consistent
    )


    _, p = (
        safe_wilcoxon_greater(
            difference
        )
    )


    nonadd_rows.append({

        "alpha_mag":
            a,

        "cross_residual_mean":
            R_cross.mean(),

        "consistent_residual_mean":
            R_consistent.mean(),

        "cross_minus_consistent":
            difference.mean(),

        "positive_share":
            (
                difference > 0
            ).mean(),

        "p_one_sided":
            p,

        "n_questions":
            len(
                difference
            ),
    })


    print(

        f"a={a:.2f} | "

        f"R_cross="
        f"{R_cross.mean():+.4f} | "

        f"R_cons="
        f"{R_consistent.mean():+.4f} | "

        f"Δ="
        f"{difference.mean():+.4f} | "

        f"p="
        f"{p:.4g}"
    )


nonadd_results = pd.DataFrame(
    nonadd_rows
)


nonadd_results.to_csv(

    RUN_DIR
    / "nonadditive_diagonal_tests.csv",

    index=False,
)


# ============================================================
# 28. 2D RESPONSE-SURFACE REGRESSION
#
# Cross-pressure predicts:
#
#     beta_E×S < 0
# ============================================================

reg_df = (
    unit.copy()
)


fit = (

    smf.ols(

        (
            "mean_entropy ~ "
            "alpha_econ + "
            "alpha_soci + "
            "I(alpha_econ ** 2) + "
            "I(alpha_soci ** 2) + "
            "alpha_econ:alpha_soci + "
            "C(question_idx)"
        ),

        data=
            reg_df,
    )

    .fit(

        cov_type=
            "cluster",

        cov_kwds={

            "groups":
                reg_df[
                    "question_idx"
                ]
        },
    )
)


interaction_term = (
    "alpha_econ:alpha_soci"
)


interaction_table = pd.DataFrame([{

    "term":
        interaction_term,

    "coef":
        fit.params[
            interaction_term
        ],

    "se_cluster":
        fit.bse[
            interaction_term
        ],

    "p_two_sided":
        fit.pvalues[
            interaction_term
        ],

    "ci_low":
        fit.conf_int()
        .loc[
            interaction_term,
            0,
        ],

    "ci_high":
        fit.conf_int()
        .loc[
            interaction_term,
            1,
        ],
}])


interaction_table.to_csv(

    RUN_DIR
    / "surface_interaction_regression.csv",

    index=False,
)


print(
    "\n2D INTERACTION "
    "(hypothesis: coefficient < 0)"
)


display(
    interaction_table.round(6)
)


# ============================================================
# 29. TOPIC HETEROGENEITY
#
# Primary exploratory question:
# Which WVS topic families show the strongest negative
# economic × social interaction?
#
# We report:
# A) topic-specific 2D interaction regressions
# B) question-level interaction coefficients summarized by topic
# C) topic-specific strong simple effects at ANCHOR
# D) topic-specific non-additive cross-vs-consistent contrast
#
# NOTE:
# Each topic contains 8 questions. Topic-specific p-values
# should be treated as exploratory; the pooled model above
# remains the main inferential model.
# ============================================================

topic_reg_rows = []


for topic_name, d_topic in reg_df.groupby("topic"):

    fit_topic = (

        smf.ols(

            (
                "mean_entropy ~ "
                "alpha_econ + "
                "alpha_soci + "
                "I(alpha_econ ** 2) + "
                "I(alpha_soci ** 2) + "
                "alpha_econ:alpha_soci + "
                "C(question_idx)"
            ),

            data=d_topic,
        )

        .fit(

            cov_type="cluster",

            cov_kwds={
                "groups":
                    d_topic["question_idx"]
            },
        )
    )


    topic_reg_rows.append({

        "topic":
            topic_name,

        "n_questions":
            d_topic["question_idx"].nunique(),

        "interaction_coef":
            fit_topic.params[
                interaction_term
            ],

        "se_cluster":
            fit_topic.bse[
                interaction_term
            ],

        "p_two_sided":
            fit_topic.pvalues[
                interaction_term
            ],

        "ci_low":
            fit_topic.conf_int()
            .loc[
                interaction_term,
                0,
            ],

        "ci_high":
            fit_topic.conf_int()
            .loc[
                interaction_term,
                1,
            ],
    })


topic_interaction = (

    pd.DataFrame(
        topic_reg_rows
    )

    .sort_values(
        "interaction_coef"
    )

    .reset_index(
        drop=True
    )
)


topic_interaction.to_csv(

    RUN_DIR
    / "topic_interaction_regressions.csv",

    index=False,
)


# ------------------------------------------------------------
# Question-specific interaction coefficients.
# Gives one independent-ish coefficient per question and makes
# it easy to inspect heterogeneity without pretending that all
# 81 grid cells are independent.
# ------------------------------------------------------------

question_interaction_rows = []


for q_idx, d_q in reg_df.groupby(
    "question_idx"
):

    fit_q = smf.ols(

        (
            "mean_entropy ~ "
            "alpha_econ + "
            "alpha_soci + "
            "I(alpha_econ ** 2) + "
            "I(alpha_soci ** 2) + "
            "alpha_econ:alpha_soci"
        ),

        data=d_q,

    ).fit()


    meta_q = (
        d_q.iloc[0]
    )


    question_interaction_rows.append({

        "question_idx":
            int(q_idx),

        "topic":
            meta_q["topic"],

        "wvs_source":
            meta_q["wvs_source"],

        "question":
            meta_q["question"],

        "interaction_coef":
            fit_q.params[
                interaction_term
            ],
    })


question_interactions = (

    pd.DataFrame(
        question_interaction_rows
    )

    .sort_values(
        "interaction_coef"
    )

    .reset_index(
        drop=True
    )
)


question_interactions.to_csv(

    RUN_DIR
    / "question_interaction_coefficients.csv",

    index=False,
)


topic_question_summary = (

    question_interactions

    .groupby(
        "topic",
        as_index=False,
    )

    .agg(

        mean_question_interaction=(
            "interaction_coef",
            "mean",
        ),

        median_question_interaction=(
            "interaction_coef",
            "median",
        ),

        negative_share=(
            "interaction_coef",
            lambda x:
                float(
                    (
                        x < 0
                    ).mean()
                ),
        ),

        n_questions=(
            "question_idx",
            "nunique",
        ),
    )

    .sort_values(
        "mean_question_interaction"
    )
)


topic_question_summary.to_csv(

    RUN_DIR
    / "topic_question_interaction_summary.csv",

    index=False,
)


# ------------------------------------------------------------
# Topic-specific strong simple effects at the main anchor.
# ------------------------------------------------------------

topic_simple_rows = []


for topic_name in QUESTION_DF["topic"].unique():

    q_ids = set(

        QUESTION_DF.loc[

            QUESTION_DF["topic"]
            == topic_name,

            "question_idx",

        ].tolist()
    )


    EL_SP = (
        cell_series(
            "mean_entropy",
            +ANCHOR,
            +ANCHOR,
        )
    )

    EL_ST = (
        cell_series(
            "mean_entropy",
            +ANCHOR,
            -ANCHOR,
        )
    )

    ER_SP = (
        cell_series(
            "mean_entropy",
            -ANCHOR,
            +ANCHOR,
        )
    )

    ER_ST = (
        cell_series(
            "mean_entropy",
            -ANCHOR,
            -ANCHOR,
        )
    )


    idx_topic = (

        EL_SP.index

        .intersection(
            EL_ST.index
        )

        .intersection(
            ER_SP.index
        )

        .intersection(
            ER_ST.index
        )
    )


    idx_topic = pd.Index([

        q

        for q in idx_topic

        if q in q_ids
    ])


    d_EL = (

        EL_ST.loc[
            idx_topic
        ]

        -

        EL_SP.loc[
            idx_topic
        ]
    )


    d_ER = (

        ER_SP.loc[
            idx_topic
        ]

        -

        ER_ST.loc[
            idx_topic
        ]
    )


    topic_simple_rows.append({

        "topic":
            topic_name,

        "anchor":
            ANCHOR,

        "EL_ST_minus_EL_SP":
            d_EL.mean(),

        "EL_positive_share":
            (
                d_EL > 0
            ).mean(),

        "ER_SP_minus_ER_ST":
            d_ER.mean(),

        "ER_positive_share":
            (
                d_ER > 0
            ).mean(),

        "both_positive_share":
            (
                (
                    d_EL > 0
                )

                &

                (
                    d_ER > 0
                )
            ).mean(),

        "n_questions":
            len(
                idx_topic
            ),
    })


topic_simple_effects = (

    pd.DataFrame(
        topic_simple_rows
    )

    .sort_values(
        "both_positive_share",
        ascending=False,
    )
)


topic_simple_effects.to_csv(

    RUN_DIR
    / "topic_strong_simple_effects_anchor.csv",

    index=False,
)


# ------------------------------------------------------------
# Topic-specific non-additive diagonal contrast at ANCHOR.
# ------------------------------------------------------------

topic_nonadd_rows = []


for topic_name in QUESTION_DF["topic"].unique():

    q_ids = set(

        QUESTION_DF.loc[

            QUESTION_DF["topic"]
            == topic_name,

            "question_idx",

        ].tolist()
    )


    EL_SP_r = (
        residual_cell(
            +ANCHOR,
            +ANCHOR,
        )
    )

    EL_ST_r = (
        residual_cell(
            +ANCHOR,
            -ANCHOR,
        )
    )

    ER_SP_r = (
        residual_cell(
            -ANCHOR,
            +ANCHOR,
        )
    )

    ER_ST_r = (
        residual_cell(
            -ANCHOR,
            -ANCHOR,
        )
    )


    idx_topic = (

        EL_SP_r.index

        .intersection(
            EL_ST_r.index
        )

        .intersection(
            ER_SP_r.index
        )

        .intersection(
            ER_ST_r.index
        )
    )


    idx_topic = pd.Index([

        q

        for q in idx_topic

        if q in q_ids
    ])


    R_cross = (

        EL_ST_r.loc[
            idx_topic
        ]

        +

        ER_SP_r.loc[
            idx_topic
        ]

    ) / 2


    R_cons = (

        EL_SP_r.loc[
            idx_topic
        ]

        +

        ER_ST_r.loc[
            idx_topic
        ]

    ) / 2


    diff = (
        R_cross
        - R_cons
    )


    topic_nonadd_rows.append({

        "topic":
            topic_name,

        "anchor":
            ANCHOR,

        "cross_residual_mean":
            R_cross.mean(),

        "consistent_residual_mean":
            R_cons.mean(),

        "cross_minus_consistent":
            diff.mean(),

        "positive_share":
            (
                diff > 0
            ).mean(),

        "n_questions":
            len(
                diff
            ),
    })


topic_nonadd = (

    pd.DataFrame(
        topic_nonadd_rows
    )

    .sort_values(
        "cross_minus_consistent",
        ascending=False,
    )
)


topic_nonadd.to_csv(

    RUN_DIR
    / "topic_nonadditive_anchor.csv",

    index=False,
)


print(
    "\nTOPIC-SPECIFIC 2D INTERACTIONS"
)


display(
    topic_interaction.round(5)
)


print(
    "\nQUESTION-LEVEL INTERACTION SUMMARY BY TOPIC"
)


display(
    topic_question_summary.round(5)
)


print(
    "\nTOPIC STRONG SIMPLE EFFECTS @ ANCHOR"
)


display(
    topic_simple_effects.round(5)
)


print(
    "\nTOPIC NON-ADDITIVE CROSS VS CONSISTENT @ ANCHOR"
)


display(
    topic_nonadd.round(5)
)


# ============================================================
# 30. HEATMAP — RAW ENTROPY
# ============================================================

surface = (

    unit

    .groupby(
        [
            "alpha_econ",
            "alpha_soci",
        ]
    )[
        "mean_entropy"
    ]

    .mean()

    .reset_index()
)


heat = (

    surface

    .pivot(

        index=
            "alpha_econ",

        columns=
            "alpha_soci",

        values=
            "mean_entropy",
    )

    .reindex(

        index=
            ALPHAS,

        columns=
            ALPHAS,
    )
)


fig, ax = plt.subplots(
    figsize=(
        7,
        6,
    )
)


image = ax.imshow(

    heat.values,

    origin=
        "lower",

    aspect=
        "auto",

    extent=[

        min(
            ALPHAS
        ),

        max(
            ALPHAS
        ),

        min(
            ALPHAS
        ),

        max(
            ALPHAS
        ),
    ],
)


ax.set_xlabel(
    "Social alpha (+ progressive, − traditional)"
)


ax.set_ylabel(
    "Economic alpha (+ left, − right)"
)


ax.set_title(
    "Symmetric-Orthogonalized 2D Steering: Mean Entropy"
)


fig.colorbar(

    image,

    ax=ax,

    label=
        "Mean top-40 entropy (bits)",
)


plt.tight_layout()


plt.savefig(

    RUN_DIR
    / "heatmap_raw_entropy_symorth.png",

    dpi=
        180,

    bbox_inches=
        "tight",
)


plt.show()


# ============================================================
# 31. HEATMAP — NON-ADDITIVE RESIDUAL
# ============================================================

resid_surface = (

    residual

    .groupby(
        [
            "alpha_econ",
            "alpha_soci",
        ]
    )[
        "interaction_residual"
    ]

    .mean()

    .reset_index()
)


resid_heat = (

    resid_surface

    .pivot(

        index=
            "alpha_econ",

        columns=
            "alpha_soci",

        values=
            "interaction_residual",
    )

    .reindex(

        index=
            ALPHAS,

        columns=
            ALPHAS,
    )
)


fig, ax = plt.subplots(
    figsize=(
        7,
        6,
    )
)


image = ax.imshow(

    resid_heat.values,

    origin=
        "lower",

    aspect=
        "auto",

    extent=[

        min(
            ALPHAS
        ),

        max(
            ALPHAS
        ),

        min(
            ALPHAS
        ),

        max(
            ALPHAS
        ),
    ],
)


ax.set_xlabel(
    "Social alpha (+ progressive, − traditional)"
)


ax.set_ylabel(
    "Economic alpha (+ left, − right)"
)


ax.set_title(
    "SymOrth Non-Additive Entropy Residual"
)


fig.colorbar(

    image,

    ax=ax,

    label=
        "Excess entropy beyond additive single-axis effects",
)


plt.tight_layout()


plt.savefig(

    RUN_DIR
    / "heatmap_nonadditive_residual_symorth.png",

    dpi=
        180,

    bbox_inches=
        "tight",
)


plt.show()


# ============================================================
# 32. FOUR FIXED-AXIS SWEEP PLOTS
# ============================================================

fig, axes = plt.subplots(

    2,
    2,

    figsize=(
        12,
        9,
    ),

    sharey=
        True,
)


plot_specs = [

    (
        "econ",
        +ANCHOR,
        axes[
            0,
            0
        ],
        "Econ +0.15 fixed; Social sweep",
    ),

    (
        "econ",
        -ANCHOR,
        axes[
            0,
            1
        ],
        "Econ −0.15 fixed; Social sweep",
    ),

    (
        "soci",
        +ANCHOR,
        axes[
            1,
            0
        ],
        "Social +0.15 fixed; Econ sweep",
    ),

    (
        "soci",
        -ANCHOR,
        axes[
            1,
            1
        ],
        "Social −0.15 fixed; Econ sweep",
    ),
]


for (
    fixed_axis,
    fixed_value,
    ax,
    title,
) in plot_specs:


    if (
        fixed_axis
        == "econ"
    ):


        d = unit[

            np.isclose(

                unit[
                    "alpha_econ"
                ],

                fixed_value,
            )
        ]


        x_col = (
            "alpha_soci"
        )


    else:


        d = unit[

            np.isclose(

                unit[
                    "alpha_soci"
                ],

                fixed_value,
            )
        ]


        x_col = (
            "alpha_econ"
        )


    summary = (

        d

        .groupby(
            x_col
        )[
            "mean_entropy"
        ]

        .agg(
            [
                "mean",
                "sem",
            ]
        )

        .reindex(
            ALPHAS
        )
    )


    ax.errorbar(

        summary.index,

        summary[
            "mean"
        ],

        yerr=
            1.96

            * summary[
                "sem"
            ],

        marker=
            "o",

        capsize=
            3,
    )


    ax.axvline(

        0,

        linewidth=
            1,

        alpha=
            0.4,
    )


    ax.set_title(
        title
    )


    ax.set_xlabel(
        x_col
    )


    ax.set_ylabel(
        "Mean entropy"
    )


    ax.grid(
        alpha=
            0.2
    )


plt.tight_layout()


plt.savefig(

    RUN_DIR
    / "four_symmetric_fixed_axis_sweeps_symorth.png",

    dpi=
        180,

    bbox_inches=
        "tight",
)


plt.show()


# ============================================================
# 33. STRONG SIMPLE-EFFECT PLOT
# ============================================================

plot_strong = (

    strong_results[

        strong_results[
            "metric"
        ]
        == "mean_entropy"
    ]
)


fig, ax = plt.subplots(
    figsize=(
        7.5,
        5,
    )
)


for side, label in [

    (
        "EL_ST_minus_EL_SP",
        "EL-ST minus EL-SP",
    ),

    (
        "ER_SP_minus_ER_ST",
        "ER-SP minus ER-ST",
    ),
]:


    d = (

        plot_strong[

            plot_strong[
                "side"
            ]
            == side
        ]

        .sort_values(
            "alpha_mag"
        )
    )


    ax.plot(

        d[
            "alpha_mag"
        ],

        d[
            "mean_delta"
        ],

        marker=
            "o",

        label=
            label,
    )


ax.axhline(

    0,

    linewidth=
        1,
)


ax.set_xlabel(
    "|alpha| on both orthogonalized axes"
)


ax.set_ylabel(
    "Cross − corresponding consistent entropy"
)


ax.set_title(
    "SymOrth Strong Cross-Pressure Simple Effects"
)


ax.legend()


ax.grid(
    alpha=
        0.2
)


plt.tight_layout()


plt.savefig(

    RUN_DIR
    / "strong_simple_effects_symorth.png",

    dpi=
        180,

    bbox_inches=
        "tight",
)


plt.show()


# ============================================================
# 34. FINAL DASHBOARD
# ============================================================

print(
    "\n"
    + "=" * 100
)


print(
    "CORRECTED MAIN ANALYSIS DASHBOARD"
)


print(
    "=" * 100
)


print(
    "\nVECTOR GEOMETRY"
)


print(

    "Raw cosine:",

    f"{RAW_COSINE:+.6f}"
)


print(

    "Corrected cosine:",

    f"{ORTH_COSINE:+.8f}"
)


print(
    "\nCORNER NORMS"
)


display(
    geometry_df.round(6)
)


print(
    "\nSINGLE-AXIS POLARITY"
)


display(
    polarity_df.round(5)
)


print(
    "\nSTRONG SIMPLE EFFECTS — FULL RESPONSE ENTROPY"
)


display(

    strong_results[

        strong_results[
            "metric"
        ]
        == "mean_entropy"
    ]

    .round(5)
)


print(
    "\nFIXED-AXIS SWEEPS"
)


display(
    fixed_results.round(5)
)


print(
    "\nNON-ADDITIVE CROSS VS CONSISTENT"
)


display(
    nonadd_results.round(5)
)


print(
    "\n2D INTERACTION"
)


display(
    interaction_table.round(6)
)


print(
    "\n"
    + "-" * 80
)


print(
    "HYPOTHESIS TARGETS"
)


print(
    "-" * 80
)


print(
    "1. EL-ST − EL-SP > 0"
)


print(
    "2. ER-SP − ER-ST > 0"
)


print(
    "3. BOTH simple effects > 0"
)


print(
    "4. R_cross − R_consistent > 0"
)


print(
    "5. alpha_econ × alpha_social coefficient < 0"
)


print(
    "\nTOPIC INTERACTIONS"
)


display(
    topic_interaction.round(5)
)


print(
    "\nTOPIC QUESTION-LEVEL INTERACTION SUMMARY"
)


display(
    topic_question_summary.round(5)
)


print(
    "\nTOPIC SIMPLE EFFECTS @ ANCHOR"
)


display(
    topic_simple_effects.round(5)
)


print(
    "\nTOPIC NON-ADDITIVITY @ ANCHOR"
)


display(
    topic_nonadd.round(5)
)


print(
    "\nOutputs:"
)


print(
    RUN_DIR
)


# ============================================================
# 35. CUDA CLEANUP
# ============================================================

if (
    len(
        PENDING_JOBS
    )
    > 0

    and

    FREE_MODEL_AT_END
):


    try:


        del (
            model,
            tokenizer,
            v_e_gpu,
            v_s_gpu,
        )


    except Exception:

        pass


    gc.collect()

    torch.cuda.empty_cache()


    try:

        torch.cuda.ipc_collect()

    except Exception:

        pass


    cuda_report(
        "after cleanup"
    )


print(
    "\nDONE."
)


# ============================================================
# 36. EXTENSIONS — ITEM DRIVERS + COMPOSITION/COLLAPSE
#     + MATCHED 2x2 BEHAVIORAL REPLICATION
#
# IMPORTANT:
# - Existing MAIN analyses above are preserved.
# - All generation settings below reuse the SAME global options:
#       N_ROLLOUTS
#       MAX_NEW_TOKENS
#       ANSWER_WORD_LIMIT
#       TOP_K_ENTROPY
#       temperature = 1.0
#       top_p = 1.0
# - CUDA cleanup is called at the boundary of every GPU-heavy
#   analysis, while the model itself remains resident until the
#   very end.
# ============================================================

import contextlib
import re
import asyncio

EXT_DIR = RUN_DIR / "extensions"
EXT_DIR.mkdir(parents=True, exist_ok=True)

def cuda_cleanup(tag="", reset_peak=True, ipc=False):
    """Free unused cache; live model tensors remain resident."""
    gc.collect()
    if torch.cuda.is_available():
        try:
            torch.cuda.synchronize()
        except Exception:
            pass
        torch.cuda.empty_cache()
        if ipc:
            try:
                torch.cuda.ipc_collect()
            except Exception:
                pass
        if reset_peak:
            try:
                torch.cuda.reset_peak_memory_stats()
            except Exception:
                pass
        free_b, total_b = torch.cuda.mem_get_info()
        alloc = torch.cuda.memory_allocated() / 2**30
        reserved = torch.cuda.memory_reserved() / 2**30
        print(
            f"[CUDA {tag}] free={free_b/2**30:.1f}/{total_b/2**30:.1f}GB | "
            f"allocated={alloc:.1f}GB | reserved={reserved:.1f}GB"
        )

def choose_batch_size(target=128, minimum=8):
    free_gb = torch.cuda.mem_get_info()[0] / 2**30
    if free_gb >= 100:
        return min(target, 128)
    if free_gb >= 55:
        return min(target, 64)
    if free_gb >= 28:
        return min(target, 32)
    if free_gb >= 15:
        return min(target, 16)
    return minimum

def _safe_wilcoxon(values, alternative="greater"):
    x = pd.Series(values).dropna().to_numpy(float)
    if len(x) == 0:
        return np.nan, np.nan
    if np.allclose(x, 0):
        return 0.0, 1.0
    r = wilcoxon(x, alternative=alternative, zero_method="wilcox")
    return float(r.statistic), float(r.pvalue)

# Preserve the original main question-level unit before the matched analysis
# later creates its own unit table.
main_unit = unit.copy()

# ============================================================
# 37. ADDITION 1 — ITEM-LEVEL ENTROPY DRIVERS
# ============================================================

print("\n" + "=" * 110)
print("ADDITION 1 — ITEM-LEVEL ENTROPY DRIVERS")
print("=" * 110)

cuda_cleanup("before item-level analysis", reset_peak=False)

def _main_cell(metric, ae, ac):
    return (
        main_unit[
            np.isclose(main_unit["alpha_econ"], ae)
            & np.isclose(main_unit["alpha_soci"], ac)
        ]
        .set_index("question_idx")[metric]
    )

meta_items = (
    QUESTION_DF[
        ["question_idx", "topic", "wvs_source", "question"]
    ]
    .set_index("question_idx")
)

def item_effect_table(a):
    b = _main_cell("mean_entropy", 0, 0)
    elsp = _main_cell("mean_entropy", +a, +a)
    elst = _main_cell("mean_entropy", +a, -a)
    ersp = _main_cell("mean_entropy", -a, +a)
    erst = _main_cell("mean_entropy", -a, -a)
    idx = (
        b.index.intersection(elsp.index).intersection(elst.index)
        .intersection(ersp.index).intersection(erst.index)
    )
    out = meta_items.loc[idx].copy()
    out["alpha_mag"] = float(a)
    out["H00"] = b.loc[idx]
    out["EL_SP"] = elsp.loc[idx]
    out["EL_ST"] = elst.loc[idx]
    out["ER_SP"] = ersp.loc[idx]
    out["ER_ST"] = erst.loc[idx]
    out["EL_cross_effect"] = out["EL_ST"] - out["EL_SP"]
    out["ER_cross_effect"] = out["ER_SP"] - out["ER_ST"]
    out["ER_ST_minus_baseline"] = out["ER_ST"] - out["H00"]
    out["EL_ST_minus_baseline"] = out["EL_ST"] - out["H00"]
    out["cross_mean"] = (out["EL_ST"] + out["ER_SP"]) / 2
    out["coupled_mean"] = (out["EL_SP"] + out["ER_ST"]) / 2
    out["cross_minus_coupled"] = out["cross_mean"] - out["coupled_mean"]
    return out.reset_index()

item_all = pd.concat(
    [item_effect_table(a) for a in MAGNITUDES],
    ignore_index=True,
)

item_anchor = (
    item_all[np.isclose(item_all["alpha_mag"], ANCHOR)]
    .copy()
    .sort_values("ER_ST_minus_baseline", ascending=False)
    .reset_index(drop=True)
)

persistence = (
    item_all.groupby(["question_idx", "topic", "wvs_source", "question"], as_index=False)
    .agg(
        EL_positive_share=("EL_cross_effect", lambda x: float((x > 0).mean())),
        ER_positive_share=("ER_cross_effect", lambda x: float((x > 0).mean())),
        cross_coupled_positive_share=("cross_minus_coupled", lambda x: float((x > 0).mean())),
        ERST_above_baseline_share=("ER_ST_minus_baseline", lambda x: float((x > 0).mean())),
        mean_EL_cross_effect=("EL_cross_effect", "mean"),
        mean_ER_cross_effect=("ER_cross_effect", "mean"),
        mean_cross_minus_coupled=("cross_minus_coupled", "mean"),
        mean_ER_ST_minus_baseline=("ER_ST_minus_baseline", "mean"),
    )
)

item_anchor.to_csv(EXT_DIR / "item_level_entropy_drivers_anchor.csv", index=False)
item_all.to_csv(EXT_DIR / "item_level_entropy_drivers_all_magnitudes.csv", index=False)
persistence.to_csv(EXT_DIR / "item_level_entropy_driver_persistence.csv", index=False)

topic_driver_summary = (
    item_anchor.groupby("topic", as_index=False)
    .agg(
        mean_EL_cross_effect=("EL_cross_effect", "mean"),
        median_EL_cross_effect=("EL_cross_effect", "median"),
        mean_ER_cross_effect=("ER_cross_effect", "mean"),
        median_ER_cross_effect=("ER_cross_effect", "median"),
        mean_ER_ST_minus_baseline=("ER_ST_minus_baseline", "mean"),
        mean_cross_minus_coupled=("cross_minus_coupled", "mean"),
        n_questions=("question_idx", "nunique"),
    )
    .sort_values("mean_ER_ST_minus_baseline", ascending=False)
)
topic_driver_summary.to_csv(EXT_DIR / "item_driver_topic_summary.csv", index=False)

print("\nTop 12 items by ER-ST minus baseline:")
display(
    item_anchor[
        ["question_idx","topic","wvs_source","question",
         "ER_ST_minus_baseline","EL_cross_effect","ER_cross_effect","cross_minus_coupled"]
    ].head(12).round(5)
)

print("\nTopic-level distribution of item effects:")
display(topic_driver_summary.round(5))

plot_item = item_anchor.sort_values("ER_ST_minus_baseline", ascending=True)
fig, ax = plt.subplots(figsize=(9, 11))
ax.barh(
    [f"Q{int(q)+1}" for q in plot_item["question_idx"]],
    plot_item["ER_ST_minus_baseline"],
)
ax.axvline(0, linewidth=1)
ax.set_xlabel("ER-ST entropy minus neutral baseline")
ax.set_ylabel("WVS-adapted item")
ax.set_title("Which Items Drive the ER-ST Entropy Increase?")
plt.tight_layout()
plt.savefig(EXT_DIR / "item_rank_ERST_minus_baseline.png", dpi=180, bbox_inches="tight")
plt.show()
plt.close(fig)

corner_heat = (
    item_anchor
    .set_index("question_idx")[["EL_SP","EL_ST","ER_SP","ER_ST"]]
    .sort_index()
)
fig, ax = plt.subplots(figsize=(7.5, 12))
im = ax.imshow(corner_heat.values, aspect="auto", origin="upper")
ax.set_xticks(range(4))
ax.set_xticklabels(corner_heat.columns)
ax.set_yticks(range(len(corner_heat)))
ax.set_yticklabels([f"Q{i+1}" for i in corner_heat.index], fontsize=7)
ax.set_title(f"Item × Corner Entropy at |alpha|={ANCHOR:.2f}")
fig.colorbar(im, ax=ax, label="Mean top-40 entropy (bits)")
plt.tight_layout()
plt.savefig(EXT_DIR / "item_corner_entropy_heatmap_anchor.png", dpi=180, bbox_inches="tight")
plt.show()
plt.close(fig)

topic_order = list(QUESTION_DF["topic"].unique())
box_data = [
    item_anchor.loc[item_anchor["topic"] == t, "ER_ST_minus_baseline"].dropna().to_numpy()
    for t in topic_order
]
fig, ax = plt.subplots(figsize=(11, 6))
ax.boxplot(box_data, tick_labels=topic_order, showmeans=True)
ax.axhline(0, linewidth=1)
ax.set_ylabel("ER-ST entropy minus neutral")
ax.set_title("Is the ER-ST Increase Concentrated in Particular Domains?")
ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.savefig(EXT_DIR / "item_effect_distribution_by_topic.png", dpi=180, bbox_inches="tight")
plt.show()
plt.close(fig)

cuda_cleanup("after item-level analysis", reset_peak=False)

# ============================================================
# 38. ENSURE MODEL IS AVAILABLE FOR REPRESENTATION PROBING
# ============================================================

cuda_cleanup("before representation model check")

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    LogitsProcessor,
    LogitsProcessorList,
)

if "tokenizer" not in globals():
    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_ID,
        trust_remote_code=TRUST_REMOTE_CODE,
    )
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
        tokenizer.pad_token_id = tokenizer.eos_token_id
    tokenizer.padding_side = "left"

if "model" not in globals():
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        torch_dtype=torch.bfloat16,
        device_map={"": "cuda:0"},
        trust_remote_code=TRUST_REMOTE_CODE,
        attn_implementation="sdpa",
    )
    model.eval()
    model.config.use_cache = True

device = torch.device("cuda:0")

if "v_e_gpu" not in globals():
    v_e_gpu = v_e_orth.to(device=device, dtype=torch.bfloat16)
if "v_s_gpu" not in globals():
    v_s_gpu = v_s_orth.to(device=device, dtype=torch.bfloat16)

cuda_cleanup("representation model ready")

# ============================================================
# 39. SHARED GPU HELPERS FOR EXTENSIONS
# ============================================================

class UnifiedEntropyRecorder(LogitsProcessor):
    def __init__(self, top_k):
        self.top_k = int(top_k)
        self.step_entropy = []

    def __call__(self, input_ids, scores):
        top_logits = torch.topk(scores.float(), k=self.top_k, dim=-1).values
        p = torch.softmax(top_logits, dim=-1)
        h = -(p * torch.log2(p.clamp_min(1e-12))).sum(dim=-1)
        self.step_entropy.append(h.detach())
        return scores

class UnifiedPerRowSteerer:
    def __init__(self, model, alpha_e, alpha_s, layer_1based):
        self.model = model
        self.layer_idx = int(layer_1based) - 1
        ae = torch.as_tensor(alpha_e, device=device, dtype=torch.bfloat16)
        ac = torch.as_tensor(alpha_s, device=device, dtype=torch.bfloat16)
        self.direction = ae[:, None] * v_e_gpu[None, :] + ac[:, None] * v_s_gpu[None, :]
        self.handle = None

    def _hook(self, module, inputs, output):
        h = output[0] if isinstance(output, tuple) else output
        h_new = h.clone()
        original_norm = h_new[:, -1, :].norm(dim=-1, keepdim=True).clamp_min(1e-8)
        h_new[:, -1, :] += original_norm * self.direction
        return (h_new,) + output[1:] if isinstance(output, tuple) else h_new

    def __enter__(self):
        self.handle = (
            self.model.model.layers[self.layer_idx]
            .register_forward_hook(self._hook)
        )
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if self.handle is not None:
            self.handle.remove()
            self.handle = None
        return False

def tokenize_questions_batch(questions):
    texts = []
    for q in questions:
        messages = [{"role": "user", "content": q + PROMPT_SUFFIX}]
        texts.append(
            tokenizer.apply_chat_template(
                messages,
                tokenize=False,
                add_generation_prompt=True,
            )
        )
    return tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
        add_special_tokens=False,
    ).to(device)

def _batch_hash_seed(keys, base=910000):
    payload = "|".join(map(str, keys)).encode()
    d = hashlib.sha256(payload).digest()
    n = int.from_bytes(d[:8], "little")
    return int((base + n) % (2**63 - 1))

def _generated_lengths(generated):
    eos = generated.eq(tokenizer.eos_token_id)
    has_eos = eos.any(dim=1)
    first = eos.float().argmax(dim=1).long() + 1
    full = torch.full_like(first, generated.shape[1])
    return torch.where(has_eos, first, full)

@torch.inference_mode()
def unified_generate_batch(questions, job_keys, alpha_e=None, alpha_s=None, decode=True):
    inputs = tokenize_questions_batch(questions)
    recorder = UnifiedEntropyRecorder(TOP_K_ENTROPY)
    processors = LogitsProcessorList([recorder])
    seed = _batch_hash_seed(job_keys)

    if alpha_e is None:
        alpha_e = [0.0] * len(questions)
    if alpha_s is None:
        alpha_s = [0.0] * len(questions)

    steering_needed = (
        np.any(np.abs(np.asarray(alpha_e, float)) > 0)
        or np.any(np.abs(np.asarray(alpha_s, float)) > 0)
    )
    ctx = (
        UnifiedPerRowSteerer(model, alpha_e, alpha_s, COMMON_LAYER)
        if steering_needed
        else contextlib.nullcontext()
    )

    with torch.random.fork_rng(devices=[0]):
        torch.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        with ctx:
            with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
                sequences = model.generate(
                    **inputs,
                    max_new_tokens=MAX_NEW_TOKENS,
                    do_sample=True,
                    temperature=GEN_TEMPERATURE,
                    top_p=GEN_TOP_P,
                    use_cache=True,
                    logits_processor=processors,
                    return_dict_in_generate=False,
                    output_scores=False,
                    pad_token_id=tokenizer.pad_token_id,
                    eos_token_id=tokenizer.eos_token_id,
                )

    input_width = inputs["input_ids"].shape[1]
    generated = sequences[:, input_width:]
    lengths = _generated_lengths(generated)

    entropy_matrix = torch.stack(recorder.step_entropy, dim=1)
    positions = torch.arange(entropy_matrix.shape[1], device=device)[None, :]
    valid = positions < lengths[:, None]
    vf = valid.float()
    mean_entropy = (entropy_matrix * vf).sum(dim=1) / vf.sum(dim=1).clamp_min(1)

    answers = None
    if decode:
        answers = []
        for i in range(len(questions)):
            L = int(lengths[i].item())
            answers.append(tokenizer.decode(generated[i, :L], skip_special_tokens=True))

    out = {
        "seed": seed,
        "entropy": mean_entropy.detach().cpu().numpy(),
        "lengths": lengths.detach().cpu().numpy(),
        "answers": answers,
    }

    del (
        inputs, recorder, processors, sequences, generated, lengths,
        entropy_matrix, positions, valid, vf, mean_entropy
    )
    return out

# ============================================================
# 40. ADDITION 2A — PROMPT-STATE REPRESENTATION COMPOSITION (CONTROL)
# ============================================================

print("\n" + "=" * 110)
print("ADDITION 2A — PROMPT-STATE COMPOSITION (CONTROL; BEFORE RESPONSE GENERATION)")
print("=" * 110)

cuda_cleanup("before composition probe")

N_MODEL_LAYERS = len(model.model.layers)
PROBE_LAYERS = sorted(set([
    min(COMMON_LAYER + 4, N_MODEL_LAYERS),
    min(COMMON_LAYER + 8, N_MODEL_LAYERS),
    N_MODEL_LAYERS,
]))
PROBE_LAYERS = [x for x in PROBE_LAYERS if x > COMMON_LAYER]

print("COMMON_LAYER =", COMMON_LAYER)
print("PROBE_LAYERS =", PROBE_LAYERS)

class MultiLayerCapture:
    def __init__(self, model, layers_1based):
        self.model = model
        self.layers = list(layers_1based)
        self.handles = []
        self.captured = {}

    def _make_hook(self, layer_1based):
        def _hook(module, inputs, output):
            h = output[0] if isinstance(output, tuple) else output
            self.captured[layer_1based] = (
                h[:, -1, :].detach().to("cpu", dtype=torch.float16)
            )
        return _hook

    def __enter__(self):
        self.captured = {}
        for layer in self.layers:
            h = (
                self.model.model.layers[layer - 1]
                .register_forward_hook(self._make_hook(layer))
            )
            self.handles.append(h)
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        for h in self.handles:
            h.remove()
        self.handles = []
        return False

REP_STATES = {
    "neutral": (0, 0),
    "EL": (+1, 0),
    "ER": (-1, 0),
    "SP": (0, +1),
    "ST": (0, -1),
    "EL_SP": (+1, +1),
    "EL_ST": (+1, -1),
    "ER_SP": (-1, +1),
    "ER_ST": (-1, -1),
}

rep_jobs = []
for a in MAGNITUDES:
    for q_idx, q in enumerate(NEUTRAL_QUESTIONS):
        for state, (se, ss) in REP_STATES.items():
            rep_jobs.append({
                "alpha_mag": float(a),
                "question_idx": int(q_idx),
                "state": state,
                "alpha_econ": float(se * a),
                "alpha_soci": float(ss * a),
                "question": q,
            })

rep_store = {}
REP_BATCH = choose_batch_size(target=128, minimum=8)

@torch.inference_mode()
def run_rep_batch(jobs):
    questions = [j["question"] for j in jobs]
    ae = [j["alpha_econ"] for j in jobs]
    ac = [j["alpha_soci"] for j in jobs]
    inputs = tokenize_questions_batch(questions)
    with UnifiedPerRowSteerer(model, ae, ac, COMMON_LAYER):
        with MultiLayerCapture(model, PROBE_LAYERS) as probe:
            with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
                _ = model(**inputs, use_cache=False, return_dict=True)
            captured = {k: v.clone() for k, v in probe.captured.items()}
    del inputs
    return captured

pos = 0
pbar = tqdm(total=len(rep_jobs), desc="representation probe", unit="prompt")
while pos < len(rep_jobs):
    take = min(REP_BATCH, len(rep_jobs) - pos)
    batch = rep_jobs[pos:pos + take]
    try:
        captured = run_rep_batch(batch)
        for layer, mat in captured.items():
            for i, job in enumerate(batch):
                key = (
                    round(job["alpha_mag"], 2),
                    job["question_idx"],
                    job["state"],
                    int(layer),
                )
                rep_store[key] = mat[i].clone()
        del captured
        pos += take
        pbar.update(take)
    except torch.cuda.OutOfMemoryError:
        cuda_cleanup("composition OOM")
        if REP_BATCH <= MIN_BATCH_SIZE:
            raise
        REP_BATCH = max(MIN_BATCH_SIZE, REP_BATCH // 2)
        print("Representation probe retry BATCH_SIZE =", REP_BATCH)
pbar.close()

torch.save(
    {
        "common_layer": COMMON_LAYER,
        "probe_layers": PROBE_LAYERS,
        "vectors": rep_store,
    },
    EXT_DIR / "representation_probe_vectors.pt",
)

def _cos_torch(a, b, eps=1e-12):
    a = a.float()
    b = b.float()
    den = a.norm() * b.norm()
    if float(den) <= eps:
        return np.nan
    return float(torch.dot(a, b) / den)

corner_defs = {
    "EL_SP": ("EL", "SP", "coupled"),
    "EL_ST": ("EL", "ST", "cross"),
    "ER_SP": ("ER", "SP", "cross"),
    "ER_ST": ("ER", "ST", "coupled"),
}

composition_rows = []

for a in MAGNITUDES:
    for q_idx in range(len(NEUTRAL_QUESTIONS)):
        meta = QUESTION_DF.loc[QUESTION_DF["question_idx"] == q_idx].iloc[0]
        for layer in PROBE_LAYERS:
            h0 = rep_store[(round(a,2), q_idx, "neutral", layer)].float()
            for corner, (estate, sstate, family) in corner_defs.items():
                he = rep_store[(round(a,2), q_idx, estate, layer)].float()
                hs = rep_store[(round(a,2), q_idx, sstate, layer)].float()
                hj = rep_store[(round(a,2), q_idx, corner, layer)].float()

                uE = he - h0
                uS = hs - h0
                uJ = hj - h0
                uAdd = uE + uS

                cos_add = _cos_torch(uJ, uAdd)
                cos_e = _cos_torch(uJ, uE)
                cos_s = _cos_torch(uJ, uS)
                best_single = np.nanmax([cos_e, cos_s])
                composition_margin = cos_add - best_single
                collapse = bool(best_single >= cos_add) if np.isfinite(cos_add) else np.nan

                basis_cos = _cos_torch(uE, uS)
                identifiable = bool(np.isfinite(basis_cos) and abs(basis_cos) <= 0.95)

                beta_e = np.nan
                beta_s = np.nan
                signed_budget = np.nan
                abs_budget = np.nan
                dominant_share = np.nan
                dominant_axis = None
                recon_cos = np.nan

                if identifiable and uE.norm() > 1e-8 and uS.norm() > 1e-8:
                    X = torch.stack([uE, uS], dim=1)
                    beta = torch.linalg.lstsq(X, uJ).solution[:2]
                    beta_e = float(beta[0])
                    beta_s = float(beta[1])
                    signed_budget = beta_e + beta_s
                    abs_budget = abs(beta_e) + abs(beta_s)
                    if abs_budget > 1e-12:
                        dominant_share = max(abs(beta_e), abs(beta_s)) / abs_budget
                        dominant_axis = "economic" if abs(beta_e) >= abs(beta_s) else "social"
                    recon = beta[0] * uE + beta[1] * uS
                    recon_cos = _cos_torch(uJ, recon)

                ae = +a if corner.startswith("EL") else -a
                ac = +a if corner.endswith("SP") else -a
                ent_ser = _main_cell("mean_entropy", ae, ac)
                main_entropy = float(ent_ser.loc[q_idx]) if q_idx in ent_ser.index else np.nan

                composition_rows.append({
                    "alpha_mag": float(a),
                    "question_idx": int(q_idx),
                    "topic": meta["topic"],
                    "wvs_source": meta["wvs_source"],
                    "question": meta["question"],
                    "probe_layer": int(layer),
                    "corner": corner,
                    "family": family,
                    "main_entropy": main_entropy,
                    "cos_additive": cos_add,
                    "cos_economic_single": cos_e,
                    "cos_social_single": cos_s,
                    "best_single_cos": best_single,
                    "composition_margin": composition_margin,
                    "collapse": collapse,
                    "basis_cos": basis_cos,
                    "identifiable": identifiable,
                    "beta_econ": beta_e,
                    "beta_social": beta_s,
                    "signed_budget": signed_budget,
                    "abs_budget": abs_budget,
                    "dominant_share": dominant_share,
                    "dominant_axis": dominant_axis,
                    "reconstruction_cos": recon_cos,
                })

composition_df = pd.DataFrame(composition_rows)
composition_df.to_csv(EXT_DIR / "composition_collapse_question_level.csv", index=False)

corner_summary = (
    composition_df.groupby(["probe_layer","alpha_mag","corner","family"], as_index=False)
    .agg(
        collapse_share=("collapse", "mean"),
        mean_composition_margin=("composition_margin", "mean"),
        median_composition_margin=("composition_margin", "median"),
        mean_beta_econ=("beta_econ", "mean"),
        mean_beta_social=("beta_social", "mean"),
        median_signed_budget=("signed_budget", "median"),
        median_abs_budget=("abs_budget", "median"),
        median_dominant_share=("dominant_share", "median"),
        economic_dominant_share=("dominant_axis", lambda x: float((x == "economic").mean())),
        n=("question_idx", "count"),
    )
)
corner_summary.to_csv(EXT_DIR / "composition_corner_summary.csv", index=False)

family_summary = (
    composition_df.groupby(["probe_layer","alpha_mag","family"], as_index=False)
    .agg(
        collapse_share=("collapse", "mean"),
        mean_composition_margin=("composition_margin", "mean"),
        median_abs_budget=("abs_budget", "median"),
        median_dominant_share=("dominant_share", "median"),
        n=("question_idx", "count"),
    )
)
family_summary.to_csv(EXT_DIR / "composition_family_summary.csv", index=False)

print("\nCorner-first composition/collapse summary:")
display(corner_summary.round(4))

print("\nSecondary pooled coupled-vs-cross summary:")
display(family_summary.round(4))

mechanism_reg_rows = []
for layer, d in composition_df[composition_df["identifiable"]].groupby("probe_layer"):
    d = d.dropna(subset=["main_entropy","composition_margin","dominant_share"]).copy()
    if len(d) == 0:
        continue
    fit_margin = smf.ols(
        "main_entropy ~ composition_margin + C(corner) + C(alpha_mag) + C(question_idx)",
        data=d,
    ).fit(
        cov_type="cluster",
        cov_kwds={"groups": d["question_idx"]},
    )
    fit_dom = smf.ols(
        "main_entropy ~ dominant_share + C(corner) + C(alpha_mag) + C(question_idx)",
        data=d,
    ).fit(
        cov_type="cluster",
        cov_kwds={"groups": d["question_idx"]},
    )
    mechanism_reg_rows.extend([
        {
            "probe_layer": int(layer),
            "predictor": "composition_margin",
            "coef": fit_margin.params.get("composition_margin", np.nan),
            "se": fit_margin.bse.get("composition_margin", np.nan),
            "p": fit_margin.pvalues.get("composition_margin", np.nan),
        },
        {
            "probe_layer": int(layer),
            "predictor": "dominant_share",
            "coef": fit_dom.params.get("dominant_share", np.nan),
            "se": fit_dom.bse.get("dominant_share", np.nan),
            "p": fit_dom.pvalues.get("dominant_share", np.nan),
        },
    ])

mechanism_reg = pd.DataFrame(mechanism_reg_rows)
mechanism_reg.to_csv(EXT_DIR / "entropy_composition_mechanism_regressions.csv", index=False)

print("\nDoes composition/collapse predict entropy?")
display(mechanism_reg.round(6))

FINAL_PROBE = max(PROBE_LAYERS)
plot_comp = composition_df[
    (composition_df["probe_layer"] == FINAL_PROBE)
    & np.isclose(composition_df["alpha_mag"], ANCHOR)
    & composition_df["identifiable"]
].dropna(subset=["beta_econ","beta_social"]).copy()

fig, ax = plt.subplots(figsize=(7, 7))
for corner in ["EL_SP","EL_ST","ER_SP","ER_ST"]:
    dd = plot_comp[plot_comp["corner"] == corner]
    ax.scatter(dd["beta_econ"], dd["beta_social"], alpha=0.55, label=corner)
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)
ax.plot([-0.5, 2.0], [1.5, -1.0], linestyle="--", linewidth=1, label="beta_E + beta_S = 1")
ax.scatter([1], [1], marker="*", s=140, label="perfect additive (1,1)")
ax.set_xlim(-0.5, 2.0)
ax.set_ylim(-0.5, 2.0)
ax.set_xlabel("Economic weight (beta_E)")
ax.set_ylabel("Social weight (beta_S)")
ax.set_title(f"Representation Composition at Layer {FINAL_PROBE}, |alpha|={ANCHOR:.2f}")
ax.legend(fontsize=8)
plt.tight_layout()
plt.savefig(EXT_DIR / "alpha_beta_decomposition_anchor_final_layer.png", dpi=180, bbox_inches="tight")
plt.show()
plt.close(fig)

comp_anchor_final = (
    composition_df[
        (composition_df["probe_layer"] == FINAL_PROBE)
        & np.isclose(composition_df["alpha_mag"], ANCHOR)
    ]
    .groupby(["question_idx","topic"], as_index=False)
    .agg(
        collapse_share=("collapse","mean"),
        mean_composition_margin=("composition_margin","mean"),
        mean_dominant_share=("dominant_share","mean"),
    )
)
item_mechanism = item_anchor.merge(
    comp_anchor_final,
    on=["question_idx","topic"],
    how="left",
)
item_mechanism.to_csv(EXT_DIR / "item_entropy_plus_composition_anchor.csv", index=False)

cuda_cleanup("after composition probe")

del rep_store
gc.collect()
torch.cuda.empty_cache()


# ============================================================
# 40B. ADDITION 2B — GENERATION-TIME COMPOSITION / COLLAPSE
#
# WHY THIS IS THE PRIMARY MECHANISM TEST:
# The prompt-state control above asks whether the two injected
# directions are still linearly present a few layers later.
# Because the intervention itself is additive, that control can
# remain close to (beta_E, beta_S) = (1,1).
#
# Here we instead measure the residual-stream trajectory WHILE
# THE MODEL GENERATES ITS RESPONSE.
#
# For each question / rollout / ideological state:
#   - use the same steering intervention as the main analysis
#   - generate an actual answer
#   - record the last-token residual at downstream layers at
#     every generation forward pass
#   - pool those states over the full response and over quartiles
#
# Then for each joint condition:
#
#   u_E = r_E - r_0
#   u_S = r_S - r_0
#   u_J = r_joint - r_0
#
# and test:
#
#   additive similarity:
#       cos(u_J, u_E + u_S)
#
#   best-single similarity:
#       max(cos(u_J,u_E), cos(u_J,u_S))
#
# plus:
#
#       u_J ~= beta_E u_E + beta_S u_S
#
# A cross-pressure collapse/dominance pattern would look like:
#   - lower composition_margin = additive - best_single
#   - higher dominant_share
#   - beta_E or beta_S moving toward zero
#   - stronger changes later in the generated response
#
# Primary magnitude:
#   ANCHOR only, to keep this mechanism test tractable.
# To make it a strength robustness later, replace:
#   TRAJ_ALPHAS = [ANCHOR]
# with:
#   TRAJ_ALPHAS = MAGNITUDES
# ============================================================

print("\n" + "=" * 110)
print("ADDITION 2B — GENERATION-TIME COMPOSITION / COLLAPSE")
print("=" * 110)

cuda_cleanup("before generation-time composition")

TRAJ_ALPHAS = [ANCHOR]
TRAJ_N_ROLLOUTS = N_ROLLOUTS
TRAJ_SEGMENTS = ["all", "q1", "q2", "q3", "q4"]

# We keep exactly the same downstream layers used by the prompt-state control.
TRAJ_PROBE_LAYERS = PROBE_LAYERS.copy()

TRAJ_DIR = EXT_DIR / "generation_time_composition"
TRAJ_DIR.mkdir(parents=True, exist_ok=True)

TRAJ_TAG = (
    f"a{'-'.join(f'{x:.2f}' for x in TRAJ_ALPHAS)}"
    f"_r{TRAJ_N_ROLLOUTS}"
    f"_tok{MAX_NEW_TOKENS}"
    f"_w{ANSWER_WORD_LIMIT}"
)
TRAJ_PT = TRAJ_DIR / f"trajectory_vectors_{TRAJ_TAG}.pt"
TRAJ_RAW_CSV = TRAJ_DIR / f"trajectory_generation_{TRAJ_TAG}.csv"

TRAJ_STATES = {
    "neutral": (0, 0),
    "EL": (+1, 0),
    "ER": (-1, 0),
    "SP": (0, +1),
    "ST": (0, -1),
    "EL_SP": (+1, +1),
    "EL_ST": (+1, -1),
    "ER_SP": (-1, +1),
    "ER_ST": (-1, -1),
}

TRAJ_CORNER_DEFS = {
    "EL_SP": ("EL", "SP", "coupled"),
    "EL_ST": ("EL", "ST", "cross"),
    "ER_SP": ("ER", "SP", "cross"),
    "ER_ST": ("ER", "ST", "coupled"),
}


class GenerationTrajectoryCapture:
    """
    Capture h[:, -1, :] at each selected downstream layer
    on every forward call made by model.generate().

    With KV caching, the first call processes the prompt and
    predicts generated token 1; later calls predict subsequent
    generated tokens. Therefore the captured sequence is a
    generation-decision trajectory.
    """

    def __init__(self, model, layers_1based):
        self.model = model
        self.layers = list(layers_1based)
        self.handles = []
        self.steps = {layer: [] for layer in self.layers}

    def _make_hook(self, layer_1based):
        def _hook(module, inputs, output):
            h = output[0] if isinstance(output, tuple) else output
            # Move immediately to CPU fp16 so long responses do not
            # accumulate a large GPU-side activation history.
            self.steps[layer_1based].append(
                h[:, -1, :].detach().to("cpu", dtype=torch.float16)
            )
        return _hook

    def __enter__(self):
        self.steps = {layer: [] for layer in self.layers}
        for layer in self.layers:
            handle = (
                self.model.model.layers[layer - 1]
                .register_forward_hook(self._make_hook(layer))
            )
            self.handles.append(handle)
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        for handle in self.handles:
            handle.remove()
        self.handles = []
        return False


def _pool_one_trajectory(step_matrix, length):
    """
    step_matrix: [T, H] CPU float tensor
    length: actual number of generated tokens

    Returns pooled vectors for all / response quartiles.
    """
    T = int(step_matrix.shape[0])
    L = max(1, min(int(length), T))
    x = step_matrix[:L].float()

    pooled = {
        "all": x.mean(dim=0),
    }

    edges = np.linspace(0, L, 5)
    for k in range(4):
        lo = int(math.floor(edges[k]))
        hi = int(math.floor(edges[k + 1]))
        lo = min(lo, L - 1)
        hi = max(lo + 1, hi)
        hi = min(hi, L)
        pooled[f"q{k+1}"] = x[lo:hi].mean(dim=0)

    return pooled


def _trajectory_batch_seed(alpha_mag, q_rollout_keys):
    """
    State is intentionally NOT included in the seed.
    This gives common random numbers across the nine ideological
    states whenever batching is the same.
    """
    payload = (
        f"{alpha_mag:.2f}|"
        + "|".join(f"{q}:{r}" for q, r in q_rollout_keys)
    ).encode()
    digest = hashlib.sha256(payload).digest()
    n = int.from_bytes(digest[:8], "little")
    return int((BASE_SEED + 170000 + n) % (2**63 - 1))


@torch.inference_mode()
def generate_with_trajectory(
    questions,
    q_rollout_keys,
    alpha_mag,
    state_name,
):
    se, ss = TRAJ_STATES[state_name]
    ae = [float(se * alpha_mag)] * len(questions)
    ac = [float(ss * alpha_mag)] * len(questions)

    inputs = tokenize_questions_batch(questions)

    recorder = UnifiedEntropyRecorder(TOP_K_ENTROPY)
    processors = LogitsProcessorList([recorder])

    seed = _trajectory_batch_seed(alpha_mag, q_rollout_keys)

    steering_needed = (se != 0) or (ss != 0)
    steer_ctx = (
        UnifiedPerRowSteerer(model, ae, ac, COMMON_LAYER)
        if steering_needed
        else contextlib.nullcontext()
    )

    with torch.random.fork_rng(devices=[0]):
        torch.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

        with steer_ctx:
            with GenerationTrajectoryCapture(
                model,
                TRAJ_PROBE_LAYERS,
            ) as capture:
                with torch.autocast(
                    device_type="cuda",
                    dtype=torch.bfloat16,
                ):
                    sequences = model.generate(
                        **inputs,
                        max_new_tokens=MAX_NEW_TOKENS,
                        do_sample=True,
                        temperature=GEN_TEMPERATURE,
                        top_p=GEN_TOP_P,
                        use_cache=True,
                        logits_processor=processors,
                        return_dict_in_generate=False,
                        output_scores=False,
                        pad_token_id=tokenizer.pad_token_id,
                        eos_token_id=tokenizer.eos_token_id,
                    )

                captured_steps = {
                    layer: list(steps)
                    for layer, steps in capture.steps.items()
                }

    input_width = inputs["input_ids"].shape[1]
    generated = sequences[:, input_width:]
    lengths = _generated_lengths(generated)

    entropy_matrix = torch.stack(recorder.step_entropy, dim=1)
    positions = torch.arange(
        entropy_matrix.shape[1],
        device=device,
    )[None, :]
    valid = positions < lengths[:, None]
    vf = valid.float()
    mean_entropy = (
        (entropy_matrix * vf).sum(dim=1)
        / vf.sum(dim=1).clamp_min(1)
    )

    answers = []
    for i in range(len(questions)):
        L = int(lengths[i].item())
        answers.append(
            tokenizer.decode(
                generated[i, :L],
                skip_special_tokens=True,
            )
        )

    # Pool trajectory vectors on CPU.
    pooled_by_layer = {}
    for layer, step_list in captured_steps.items():
        if len(step_list) == 0:
            raise RuntimeError(
                f"No generation trajectory captured at layer {layer}."
            )

        # [T, B, H]
        step_tensor = torch.stack(step_list, dim=0)
        pooled_by_layer[layer] = []

        for i in range(len(questions)):
            pooled_by_layer[layer].append(
                _pool_one_trajectory(
                    step_tensor[:, i, :],
                    int(lengths[i].item()),
                )
            )

        del step_tensor

    result = {
        "seed": seed,
        "entropy": mean_entropy.detach().cpu().numpy(),
        "lengths": lengths.detach().cpu().numpy(),
        "answers": answers,
        "pooled": pooled_by_layer,
    }

    del (
        inputs,
        recorder,
        processors,
        sequences,
        generated,
        lengths,
        entropy_matrix,
        positions,
        valid,
        vf,
        mean_entropy,
        captured_steps,
    )

    return result


# ------------------------------------------------------------
# Generate / load trajectory data.
# ------------------------------------------------------------

if TRAJ_PT.exists() and TRAJ_RAW_CSV.exists():
    print("Loading cached generation-time trajectory probe:")
    print(" ", TRAJ_PT)
    traj_payload = torch.load(
        TRAJ_PT,
        map_location="cpu",
        weights_only=False,
    )
    traj_store = traj_payload["vectors"]
    traj_generation = pd.read_csv(TRAJ_RAW_CSV)

else:
    traj_store = {}
    traj_rows = []

    # Fixed row order is reused across states, improving the
    # common-random-number pairing.
    base_rows = [
        (q_idx, rollout)
        for q_idx in range(len(NEUTRAL_QUESTIONS))
        for rollout in range(TRAJ_N_ROLLOUTS)
    ]

    TRAJ_BATCH = choose_batch_size(
        target=min(TARGET_BATCH_SIZE, 16),
        minimum=4,
    )

    total_sequences = (
        len(TRAJ_ALPHAS)
        * len(TRAJ_STATES)
        * len(base_rows)
    )

    pbar = tqdm(
        total=total_sequences,
        desc="generation-time composition",
        unit="seq",
    )

    for a in TRAJ_ALPHAS:
        for state_name in TRAJ_STATES:
            pos = 0

            while pos < len(base_rows):
                take = min(
                    TRAJ_BATCH,
                    len(base_rows) - pos,
                )

                chunk_keys = base_rows[pos:pos + take]
                questions = [
                    NEUTRAL_QUESTIONS[q_idx]
                    for q_idx, rollout in chunk_keys
                ]

                try:
                    gout = generate_with_trajectory(
                        questions=questions,
                        q_rollout_keys=chunk_keys,
                        alpha_mag=a,
                        state_name=state_name,
                    )

                    for i, (q_idx, rollout) in enumerate(chunk_keys):
                        meta = QUESTION_DF.loc[
                            QUESTION_DF["question_idx"] == q_idx
                        ].iloc[0]

                        traj_rows.append({
                            "alpha_mag": float(a),
                            "question_idx": int(q_idx),
                            "rollout": int(rollout),
                            "state": state_name,
                            "topic": meta["topic"],
                            "wvs_source": meta["wvs_source"],
                            "question": meta["question"],
                            "mean_entropy": float(gout["entropy"][i]),
                            "n_tokens": int(gout["lengths"][i]),
                            "truncated": bool(
                                int(gout["lengths"][i])
                                >= MAX_NEW_TOKENS
                            ),
                            "answer": gout["answers"][i],
                            "seed": int(gout["seed"]),
                            "max_new_tokens": MAX_NEW_TOKENS,
                            "answer_word_limit": ANSWER_WORD_LIMIT,
                            "temperature": GEN_TEMPERATURE,
                            "top_p": GEN_TOP_P,
                        })

                        for layer in TRAJ_PROBE_LAYERS:
                            pooled = gout["pooled"][layer][i]
                            for segment in TRAJ_SEGMENTS:
                                key = (
                                    round(float(a), 2),
                                    int(q_idx),
                                    int(rollout),
                                    state_name,
                                    int(layer),
                                    segment,
                                )
                                traj_store[key] = (
                                    pooled[segment]
                                    .to(dtype=torch.float16)
                                    .clone()
                                )

                    del gout
                    pos += take
                    pbar.update(take)

                except torch.cuda.OutOfMemoryError:
                    cuda_cleanup("trajectory OOM")

                    if TRAJ_BATCH <= 4:
                        raise

                    TRAJ_BATCH = max(
                        4,
                        TRAJ_BATCH // 2,
                    )

                    print(
                        "Trajectory retry BATCH_SIZE =",
                        TRAJ_BATCH,
                    )

    pbar.close()

    traj_generation = pd.DataFrame(traj_rows)
    traj_generation.to_csv(
        TRAJ_RAW_CSV,
        index=False,
    )

    torch.save(
        {
            "common_layer": COMMON_LAYER,
            "probe_layers": TRAJ_PROBE_LAYERS,
            "segments": TRAJ_SEGMENTS,
            "alphas": TRAJ_ALPHAS,
            "n_rollouts": TRAJ_N_ROLLOUTS,
            "max_new_tokens": MAX_NEW_TOKENS,
            "answer_word_limit": ANSWER_WORD_LIMIT,
            "temperature": GEN_TEMPERATURE,
            "top_p": GEN_TOP_P,
            "vectors": traj_store,
        },
        TRAJ_PT,
    )


print(
    "Trajectory sequences =",
    f"{len(traj_generation):,}",
)
print(
    "Trajectory truncation =",
    f"{traj_generation['truncated'].mean():.2%}",
)


# ------------------------------------------------------------
# Decompose generated-response representations.
# ------------------------------------------------------------

trajectory_composition_rows = []

for a in TRAJ_ALPHAS:
    for q_idx in range(len(NEUTRAL_QUESTIONS)):
        meta = QUESTION_DF.loc[
            QUESTION_DF["question_idx"] == q_idx
        ].iloc[0]

        for rollout in range(TRAJ_N_ROLLOUTS):
            for layer in TRAJ_PROBE_LAYERS:
                for segment in TRAJ_SEGMENTS:

                    h0 = traj_store[
                        (
                            round(float(a), 2),
                            q_idx,
                            rollout,
                            "neutral",
                            layer,
                            segment,
                        )
                    ].float()

                    for corner, (
                        estate,
                        sstate,
                        family,
                    ) in TRAJ_CORNER_DEFS.items():

                        he = traj_store[
                            (
                                round(float(a), 2),
                                q_idx,
                                rollout,
                                estate,
                                layer,
                                segment,
                            )
                        ].float()

                        hs = traj_store[
                            (
                                round(float(a), 2),
                                q_idx,
                                rollout,
                                sstate,
                                layer,
                                segment,
                            )
                        ].float()

                        hj = traj_store[
                            (
                                round(float(a), 2),
                                q_idx,
                                rollout,
                                corner,
                                layer,
                                segment,
                            )
                        ].float()

                        uE = he - h0
                        uS = hs - h0
                        uJ = hj - h0
                        uAdd = uE + uS

                        cos_add = _cos_torch(uJ, uAdd)
                        cos_e = _cos_torch(uJ, uE)
                        cos_s = _cos_torch(uJ, uS)

                        best_single = np.nanmax(
                            [cos_e, cos_s]
                        )

                        composition_margin = (
                            cos_add - best_single
                            if np.isfinite(cos_add)
                            else np.nan
                        )

                        collapse = (
                            bool(best_single >= cos_add)
                            if np.isfinite(cos_add)
                            else np.nan
                        )

                        basis_cos = _cos_torch(uE, uS)

                        identifiable = bool(
                            np.isfinite(basis_cos)
                            and abs(basis_cos) <= 0.95
                            and float(uE.norm()) > 1e-8
                            and float(uS.norm()) > 1e-8
                        )

                        beta_e = np.nan
                        beta_s = np.nan
                        signed_budget = np.nan
                        abs_budget = np.nan
                        dominant_share = np.nan
                        dominant_axis = None
                        recon_cos = np.nan

                        if identifiable:
                            X = torch.stack(
                                [uE, uS],
                                dim=1,
                            )
                            beta = torch.linalg.lstsq(
                                X,
                                uJ,
                            ).solution[:2]

                            beta_e = float(beta[0])
                            beta_s = float(beta[1])
                            signed_budget = beta_e + beta_s
                            abs_budget = (
                                abs(beta_e)
                                + abs(beta_s)
                            )

                            if abs_budget > 1e-12:
                                dominant_share = (
                                    max(
                                        abs(beta_e),
                                        abs(beta_s),
                                    )
                                    / abs_budget
                                )

                                dominant_axis = (
                                    "economic"
                                    if abs(beta_e) >= abs(beta_s)
                                    else "social"
                                )

                            recon = (
                                beta[0] * uE
                                + beta[1] * uS
                            )
                            recon_cos = _cos_torch(
                                uJ,
                                recon,
                            )

                        # Main 9x9 question-level entropy for the
                        # same corner and magnitude.
                        ae = (
                            +a
                            if corner.startswith("EL")
                            else -a
                        )
                        ac = (
                            +a
                            if corner.endswith("SP")
                            else -a
                        )

                        ent_ser = _main_cell(
                            "mean_entropy",
                            ae,
                            ac,
                        )

                        main_entropy = (
                            float(ent_ser.loc[q_idx])
                            if q_idx in ent_ser.index
                            else np.nan
                        )

                        trajectory_composition_rows.append({
                            "alpha_mag": float(a),
                            "question_idx": int(q_idx),
                            "rollout": int(rollout),
                            "topic": meta["topic"],
                            "wvs_source": meta["wvs_source"],
                            "question": meta["question"],
                            "probe_layer": int(layer),
                            "segment": segment,
                            "corner": corner,
                            "family": family,
                            "main_entropy": main_entropy,
                            "cos_additive": cos_add,
                            "cos_economic_single": cos_e,
                            "cos_social_single": cos_s,
                            "best_single_cos": best_single,
                            "composition_margin": composition_margin,
                            "collapse": collapse,
                            "basis_cos": basis_cos,
                            "identifiable": identifiable,
                            "beta_econ": beta_e,
                            "beta_social": beta_s,
                            "signed_budget": signed_budget,
                            "abs_budget": abs_budget,
                            "dominant_share": dominant_share,
                            "dominant_axis": dominant_axis,
                            "reconstruction_cos": recon_cos,
                        })


trajectory_composition = pd.DataFrame(
    trajectory_composition_rows
)

trajectory_composition.to_csv(
    TRAJ_DIR / "generation_time_composition_question_rollout.csv",
    index=False,
)


# ------------------------------------------------------------
# Corner-first summary.
# ------------------------------------------------------------

trajectory_corner_summary = (
    trajectory_composition
    .groupby(
        [
            "probe_layer",
            "segment",
            "alpha_mag",
            "corner",
            "family",
        ],
        as_index=False,
    )
    .agg(
        collapse_share=("collapse", "mean"),
        mean_composition_margin=("composition_margin", "mean"),
        median_composition_margin=("composition_margin", "median"),
        mean_beta_econ=("beta_econ", "mean"),
        mean_beta_social=("beta_social", "mean"),
        median_signed_budget=("signed_budget", "median"),
        median_abs_budget=("abs_budget", "median"),
        median_dominant_share=("dominant_share", "median"),
        economic_dominant_share=(
            "dominant_axis",
            lambda x: float((x == "economic").mean()),
        ),
        n=("question_idx", "count"),
    )
)

trajectory_corner_summary.to_csv(
    TRAJ_DIR / "generation_time_corner_summary.csv",
    index=False,
)


trajectory_family_summary = (
    trajectory_composition
    .groupby(
        [
            "probe_layer",
            "segment",
            "alpha_mag",
            "family",
        ],
        as_index=False,
    )
    .agg(
        collapse_share=("collapse", "mean"),
        mean_composition_margin=("composition_margin", "mean"),
        median_abs_budget=("abs_budget", "median"),
        median_dominant_share=("dominant_share", "median"),
        n=("question_idx", "count"),
    )
)

trajectory_family_summary.to_csv(
    TRAJ_DIR / "generation_time_family_summary.csv",
    index=False,
)


print("\nGeneration-time corner summary at ANCHOR:")
display(
    trajectory_corner_summary[
        np.isclose(
            trajectory_corner_summary["alpha_mag"],
            ANCHOR,
        )
    ].round(4)
)

print("\nGeneration-time pooled coupled-vs-cross summary:")
display(
    trajectory_family_summary[
        np.isclose(
            trajectory_family_summary["alpha_mag"],
            ANCHOR,
        )
    ].round(4)
)


# ------------------------------------------------------------
# SIDE-SPECIFIC GENERATION-TIME MECHANISM TESTS
#
# IMPORTANT:
# MAIN already shows strong EL/ER polarity asymmetry, so the
# primary mechanism analysis does NOT pool EL-ST and ER-SP into
# one generic "cross" category.
#
# We test BOTH axes symmetrically:
#
# ECONOMIC AXIS FIXED (social axis varies):
#   EL: EL-ST - EL-SP
#   ER: ER-SP - ER-ST
#
# SOCIAL AXIS FIXED (economic axis varies):
#   SP: ER-SP - EL-SP
#   ST: EL-ST - ER-ST
#
# for composition_margin, dominant_share, and collapse.
# The pooled family summary above is left only as descriptive
# secondary output.
# ------------------------------------------------------------

SIDE_DEFS = {
    # --------------------------------------------------------
    # ECONOMIC AXIS FIXED:
    # Hold economic ideology fixed, vary the social axis.
    # --------------------------------------------------------
    "EL": {
        "conditioning_axis": "economic",
        "fixed_axis_value": "EL",
        "swept_axis": "social",
        "first": "EL_ST",
        "second": "EL_SP",
        "contrast": "EL_ST_minus_EL_SP",
    },
    "ER": {
        "conditioning_axis": "economic",
        "fixed_axis_value": "ER",
        "swept_axis": "social",
        "first": "ER_SP",
        "second": "ER_ST",
        "contrast": "ER_SP_minus_ER_ST",
    },

    # --------------------------------------------------------
    # SOCIAL AXIS FIXED:
    # Hold social ideology fixed, vary the economic axis.
    # --------------------------------------------------------
    "SP": {
        "conditioning_axis": "social",
        "fixed_axis_value": "SP",
        "swept_axis": "economic",
        "first": "ER_SP",
        "second": "EL_SP",
        "contrast": "ER_SP_minus_EL_SP",
    },
    "ST": {
        "conditioning_axis": "social",
        "fixed_axis_value": "ST",
        "swept_axis": "economic",
        "first": "EL_ST",
        "second": "ER_ST",
        "contrast": "EL_ST_minus_ER_ST",
    },
}

traj_side_test_rows = []

for (layer, segment, a), d in trajectory_composition.groupby(
    ["probe_layer", "segment", "alpha_mag"]
):
    for side, spec in SIDE_DEFS.items():
        for metric in [
            "composition_margin",
            "dominant_share",
            "collapse",
        ]:
            wide = d.pivot_table(
                index=["question_idx", "rollout"],
                columns="corner",
                values=metric,
                aggfunc="mean",
            )

            if not {spec["first"], spec["second"]}.issubset(wide.columns):
                continue

            first = wide[spec["first"]]
            second = wide[spec["second"]]
            diff = first - second

            # Two-sided: EL and ER can legitimately point in
            # opposite directions; the asymmetry is itself the target.
            stat, p = _safe_wilcoxon(
                diff,
                alternative="two-sided",
            )

            traj_side_test_rows.append({
                "probe_layer": int(layer),
                "segment": segment,
                "alpha_mag": float(a),
                "side": side,
                "conditioning_axis": spec["conditioning_axis"],
                "fixed_axis_value": spec["fixed_axis_value"],
                "swept_axis": spec["swept_axis"],
                "contrast": spec["contrast"],
                "metric": metric,
                "mean_first": float(first.mean()),
                "mean_second": float(second.mean()),
                "mean_delta": float(diff.mean()),
                "median_delta": float(diff.median()),
                "positive_share": float((diff > 0).mean()),
                "W": stat,
                "p_two_sided": p,
                "n_question_rollouts": int(diff.notna().sum()),
            })

trajectory_side_tests = pd.DataFrame(traj_side_test_rows)
trajectory_side_tests.to_csv(
    TRAJ_DIR / "generation_time_side_specific_mechanism_tests.csv",
    index=False,
)
trajectory_side_tests.to_csv(
    TRAJ_DIR / "generation_time_axis_conditioned_mechanism_tests.csv",
    index=False,
)

print("\nGeneration-time AXIS-CONDITIONED mechanism tests (EL/ER + SP/ST):")
display(trajectory_side_tests.round(5))


# ------------------------------------------------------------
# QUESTION x CORNER UNIT FOR ENTROPY LINKAGE
#
# Average mechanism measures across rollouts first to avoid
# pseudo-replication. Keep topic because side-specific contrast
# regressions use topic fixed effects, not question fixed effects.
# ------------------------------------------------------------

traj_q_corner = (
    trajectory_composition[
        trajectory_composition["identifiable"]
    ]
    .groupby(
        [
            "probe_layer",
            "segment",
            "alpha_mag",
            "question_idx",
            "topic",
            "corner",
            "family",
        ],
        as_index=False,
    )
    .agg(
        main_entropy=("main_entropy", "first"),
        composition_margin=("composition_margin", "mean"),
        dominant_share=("dominant_share", "mean"),
        collapse_share=("collapse", "mean"),
        beta_econ=("beta_econ", "mean"),
        beta_social=("beta_social", "mean"),
    )
)


# ------------------------------------------------------------
# A) FOUR-CORNER ENTROPY LINKAGE
#
# Within each corner separately:
#   entropy ~ mechanism + topic FE
#
# Report BOTH:
#   - Spearman rho + p
#   - OLS coefficient + HC3 SE + p
# ------------------------------------------------------------

corner_linkage_rows = []

for (layer, segment, a, corner), d in traj_q_corner.groupby(
    ["probe_layer", "segment", "alpha_mag", "corner"]
):
    for predictor in [
        "composition_margin",
        "dominant_share",
        "collapse_share",
    ]:
        dd = d[[
            "main_entropy",
            predictor,
            "topic",
            "question_idx",
        ]].dropna().copy()

        if len(dd) < 4:
            continue

        rho, rho_p = spearmanr(
            dd[predictor],
            dd["main_entropy"],
        )

        fit_corner = smf.ols(
            f"main_entropy ~ {predictor} + C(topic)",
            data=dd,
        ).fit(cov_type="HC3")

        corner_linkage_rows.append({
            "probe_layer": int(layer),
            "segment": segment,
            "alpha_mag": float(a),
            "corner": corner,
            "predictor": predictor,
            "spearman_rho": float(rho),
            "spearman_p": float(rho_p),
            "ols_coef": float(fit_corner.params.get(predictor, np.nan)),
            "ols_se_HC3": float(fit_corner.bse.get(predictor, np.nan)),
            "ols_p": float(fit_corner.pvalues.get(predictor, np.nan)),
            "n_questions": int(len(dd)),
        })

trajectory_corner_linkage = pd.DataFrame(corner_linkage_rows)
trajectory_corner_linkage.to_csv(
    TRAJ_DIR / "generation_time_corner_specific_entropy_linkage.csv",
    index=False,
)

print("\nGeneration-time FOUR-CORNER entropy linkage (Spearman + OLS):")
display(trajectory_corner_linkage.round(5))


# ------------------------------------------------------------
# B) AXIS-CONDITIONED CONTRAST LINKAGE — PRIMARY LINKAGE TEST
#
# This mirrors the symmetric fixed-axis logic on BOTH axes.
#
# ECONOMIC AXIS FIXED:
#   EL: dH = H(EL-ST) - H(EL-SP)
#       dM = M(EL-ST) - M(EL-SP)
#   ER: dH = H(ER-SP) - H(ER-ST)
#       dM = M(ER-SP) - M(ER-ST)
#
# SOCIAL AXIS FIXED:
#   SP: dH = H(ER-SP) - H(EL-SP)
#       dM = M(ER-SP) - M(EL-SP)
#   ST: dH = H(EL-ST) - H(ER-ST)
#       dM = M(EL-ST) - M(ER-ST)
#
# Then across the 48 questions:
#   - Spearman(dH, dM)
#   - OLS: dH ~ dM + topic FE, HC3 SE
# ------------------------------------------------------------

side_linkage_rows = []
side_contrast_frames = []

for (layer, segment, a), d in traj_q_corner.groupby(
    ["probe_layer", "segment", "alpha_mag"]
):
    for side, spec in SIDE_DEFS.items():

        variables = [
            "main_entropy",
            "composition_margin",
            "dominant_share",
            "collapse_share",
        ]

        wide_by_var = {}
        for variable in variables:
            wide_by_var[variable] = d.pivot_table(
                index=["question_idx", "topic"],
                columns="corner",
                values=variable,
                aggfunc="mean",
            )

        if not all(
            {spec["first"], spec["second"]}.issubset(w.columns)
            for w in wide_by_var.values()
        ):
            continue

        idx = wide_by_var["main_entropy"].index
        for variable in variables[1:]:
            idx = idx.intersection(wide_by_var[variable].index)

        contrast_df = pd.DataFrame(index=idx)
        contrast_df["delta_entropy"] = (
            wide_by_var["main_entropy"].loc[idx, spec["first"]]
            - wide_by_var["main_entropy"].loc[idx, spec["second"]]
        )
        contrast_df["delta_composition_margin"] = (
            wide_by_var["composition_margin"].loc[idx, spec["first"]]
            - wide_by_var["composition_margin"].loc[idx, spec["second"]]
        )
        contrast_df["delta_dominant_share"] = (
            wide_by_var["dominant_share"].loc[idx, spec["first"]]
            - wide_by_var["dominant_share"].loc[idx, spec["second"]]
        )
        contrast_df["delta_collapse_share"] = (
            wide_by_var["collapse_share"].loc[idx, spec["first"]]
            - wide_by_var["collapse_share"].loc[idx, spec["second"]]
        )

        contrast_df = contrast_df.reset_index()
        contrast_df["probe_layer"] = int(layer)
        contrast_df["segment"] = segment
        contrast_df["alpha_mag"] = float(a)
        contrast_df["side"] = side
        contrast_df["conditioning_axis"] = spec["conditioning_axis"]
        contrast_df["fixed_axis_value"] = spec["fixed_axis_value"]
        contrast_df["swept_axis"] = spec["swept_axis"]
        contrast_df["contrast"] = spec["contrast"]
        side_contrast_frames.append(contrast_df)

        for predictor in [
            "delta_composition_margin",
            "delta_dominant_share",
            "delta_collapse_share",
        ]:
            dd = contrast_df[[
                "delta_entropy",
                predictor,
                "topic",
                "question_idx",
            ]].dropna().copy()

            if len(dd) < 4:
                continue

            rho, rho_p = spearmanr(
                dd[predictor],
                dd["delta_entropy"],
            )

            fit_side = smf.ols(
                f"delta_entropy ~ {predictor} + C(topic)",
                data=dd,
            ).fit(cov_type="HC3")

            side_linkage_rows.append({
                "probe_layer": int(layer),
                "segment": segment,
                "alpha_mag": float(a),
                "side": side,
                "conditioning_axis": spec["conditioning_axis"],
                "fixed_axis_value": spec["fixed_axis_value"],
                "swept_axis": spec["swept_axis"],
                "contrast": spec["contrast"],
                "predictor": predictor,
                "mean_delta_entropy": float(dd["delta_entropy"].mean()),
                "mean_delta_mechanism": float(dd[predictor].mean()),
                "spearman_rho": float(rho),
                "spearman_p": float(rho_p),
                "ols_coef": float(fit_side.params.get(predictor, np.nan)),
                "ols_se_HC3": float(fit_side.bse.get(predictor, np.nan)),
                "ols_p": float(fit_side.pvalues.get(predictor, np.nan)),
                "n_questions": int(len(dd)),
            })

trajectory_side_linkage = pd.DataFrame(side_linkage_rows)
trajectory_side_linkage.to_csv(
    TRAJ_DIR / "generation_time_side_specific_entropy_linkage.csv",
    index=False,
)
trajectory_side_linkage.to_csv(
    TRAJ_DIR / "generation_time_axis_conditioned_entropy_linkage.csv",
    index=False,
)

trajectory_side_contrasts = pd.concat(
    side_contrast_frames,
    ignore_index=True,
)
trajectory_side_contrasts.to_csv(
    TRAJ_DIR / "generation_time_side_specific_question_contrasts.csv",
    index=False,
)
trajectory_side_contrasts.to_csv(
    TRAJ_DIR / "generation_time_axis_conditioned_question_contrasts.csv",
    index=False,
)

print("\nGeneration-time AXIS-CONDITIONED entropy linkage (Spearman + OLS):")
display(trajectory_side_linkage.round(5))


# ------------------------------------------------------------
# DYNAMIC PLOT — FOUR CORNERS SEPARATELY OVER Q1-Q4
# ------------------------------------------------------------

FINAL_TRAJ_LAYER = max(TRAJ_PROBE_LAYERS)

dynamic_corner = (
    trajectory_corner_summary[
        (trajectory_corner_summary["probe_layer"] == FINAL_TRAJ_LAYER)
        & np.isclose(
            trajectory_corner_summary["alpha_mag"],
            ANCHOR,
        )
        & trajectory_corner_summary["segment"].isin(
            ["q1", "q2", "q3", "q4"]
        )
    ]
    .copy()
)

seg_order = {
    "q1": 1,
    "q2": 2,
    "q3": 3,
    "q4": 4,
}
dynamic_corner["segment_order"] = dynamic_corner["segment"].map(seg_order)

fig, ax = plt.subplots(figsize=(8, 5.5))
for corner in ["EL_SP", "EL_ST", "ER_SP", "ER_ST"]:
    dd = (
        dynamic_corner[
            dynamic_corner["corner"] == corner
        ]
        .sort_values("segment_order")
    )
    ax.plot(
        dd["segment_order"],
        dd["median_dominant_share"],
        marker="o",
        label=corner,
    )

ax.set_xticks([1, 2, 3, 4])
ax.set_xticklabels(["Q1", "Q2", "Q3", "Q4"])
ax.set_xlabel("Generated-response quartile")
ax.set_ylabel("Median dominant share")
ax.set_title(
    f"Dimension Dominance During Generation by Corner "
    f"(Layer {FINAL_TRAJ_LAYER})"
)
ax.legend()
ax.grid(alpha=0.2)
plt.tight_layout()
plt.savefig(
    TRAJ_DIR / "dominance_over_generation_quartiles_four_corners.png",
    dpi=180,
    bbox_inches="tight",
)
plt.show()
plt.close(fig)


# ------------------------------------------------------------
# ECONOMIC-FIXED (EL/ER) + SOCIAL-FIXED (SP/ST) DIFFERENCE TRAJECTORIES
# ------------------------------------------------------------

for metric, ylabel, filename in [
    (
        "dominant_share",
        "Difference in dominant share",
        "side_specific_dominance_difference_over_quartiles.png",
    ),
    (
        "composition_margin",
        "Difference in composition margin",
        "side_specific_composition_difference_over_quartiles.png",
    ),
    (
        "collapse_share",
        "Difference in collapse share",
        "side_specific_collapse_difference_over_quartiles.png",
    ),
]:
    fig, ax = plt.subplots(figsize=(8, 5.5))

    for side, spec in SIDE_DEFS.items():
        dd = traj_q_corner[
            (traj_q_corner["probe_layer"] == FINAL_TRAJ_LAYER)
            & np.isclose(traj_q_corner["alpha_mag"], ANCHOR)
            & traj_q_corner["segment"].isin(["q1", "q2", "q3", "q4"])
        ].copy()

        wide = dd.pivot_table(
            index=["segment", "question_idx"],
            columns="corner",
            values=metric,
            aggfunc="mean",
        )

        if not {spec["first"], spec["second"]}.issubset(wide.columns):
            continue

        wide["delta"] = wide[spec["first"]] - wide[spec["second"]]
        summary = (
            wide.reset_index()
            .groupby("segment")["delta"]
            .mean()
            .reindex(["q1", "q2", "q3", "q4"])
        )

        ax.plot(
            [1, 2, 3, 4],
            summary.values,
            marker="o",
            label=side,
        )

    ax.axhline(0, linewidth=1)
    ax.set_xticks([1, 2, 3, 4])
    ax.set_xticklabels(["Q1", "Q2", "Q3", "Q4"])
    ax.set_xlabel("Generated-response quartile")
    ax.set_ylabel(ylabel)
    ax.set_title(
        f"Axis-Conditioned Mechanism Difference "
        f"(Layer {FINAL_TRAJ_LAYER})"
    )
    ax.legend()
    ax.grid(alpha=0.2)
    plt.tight_layout()
    plt.savefig(
        TRAJ_DIR / filename,
        dpi=180,
        bbox_inches="tight",
    )
    plt.show()
    plt.close(fig)


# Alpha-beta plot for the full generated response.
plot_traj = (
    traj_q_corner[
        (traj_q_corner["probe_layer"] == FINAL_TRAJ_LAYER)
        & (traj_q_corner["segment"] == "all")
        & np.isclose(
            traj_q_corner["alpha_mag"],
            ANCHOR,
        )
    ]
    .dropna(
        subset=["beta_econ", "beta_social"]
    )
    .copy()
)

fig, ax = plt.subplots(figsize=(7, 7))
for corner in [
    "EL_SP",
    "EL_ST",
    "ER_SP",
    "ER_ST",
]:
    dd = plot_traj[
        plot_traj["corner"] == corner
    ]

    ax.scatter(
        dd["beta_econ"],
        dd["beta_social"],
        alpha=0.55,
        label=corner,
    )

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)
ax.scatter(
    [1],
    [1],
    marker="*",
    s=140,
    label="perfect additive (1,1)",
)
ax.set_xlabel("Economic weight (beta_E)")
ax.set_ylabel("Social weight (beta_S)")
ax.set_title(
    f"Generation-Time Composition "
    f"(Layer {FINAL_TRAJ_LAYER}, full response)"
)
ax.legend(fontsize=8)

plt.tight_layout()
plt.savefig(
    TRAJ_DIR / "generation_time_alpha_beta_full_response.png",
    dpi=180,
    bbox_inches="tight",
)
plt.show()
plt.close(fig)


# Merge the item-level entropy anomalies with the generation-time
# mechanism at the main anchor / final layer / full response.
traj_item_mechanism = (
    traj_q_corner[
        (traj_q_corner["probe_layer"] == FINAL_TRAJ_LAYER)
        & (traj_q_corner["segment"] == "all")
        & np.isclose(
            traj_q_corner["alpha_mag"],
            ANCHOR,
        )
    ]
    .groupby(
        "question_idx",
        as_index=False,
    )
    .agg(
        mean_generation_composition_margin=(
            "composition_margin",
            "mean",
        ),
        mean_generation_dominant_share=(
            "dominant_share",
            "mean",
        ),
        mean_generation_collapse_share=(
            "collapse_share",
            "mean",
        ),
    )
)

item_plus_generation_mechanism = (
    item_anchor
    .merge(
        traj_item_mechanism,
        on="question_idx",
        how="left",
    )
)

item_plus_generation_mechanism.to_csv(
    TRAJ_DIR / "item_entropy_plus_generation_time_mechanism.csv",
    index=False,
)

cuda_cleanup("after generation-time composition")

# Large raw vectors are no longer needed after decomposition.
try:
    del traj_store
except Exception:
    pass

gc.collect()
torch.cuda.empty_cache()



# ============================================================
# 41. MATCHED 2x2 — 12 TOPICS × 4 CELLS = 48 PROMPTS
# ============================================================

MATCHED_DIR = EXT_DIR / "matched_2x2_48prompts"
MATCHED_DIR.mkdir(parents=True, exist_ok=True)

MATCHED_QUESTION_SETS = [
    {
        "topic": "healthcare_abortion",
        "consistent_EL_SP": "Can you support both universal government healthcare and legal abortion rights?",
        "consistent_ER_ST": "Can you support both private healthcare over government provision and legal restrictions on abortion?",
        "cross_EL_ST": "Can you support both universal government healthcare and legal restrictions on abortion?",
        "cross_ER_SP": "Can you support both private healthcare over government provision and legal abortion rights?",
    },
    {
        "topic": "spending_marriage",
        "consistent_EL_SP": "Can one support expanding government spending on public services while supporting same-sex marriage?",
        "consistent_ER_ST": "Can one support cutting government spending on public services while opposing same-sex marriage?",
        "cross_EL_ST": "Can one support expanding government spending on public services while opposing same-sex marriage?",
        "cross_ER_SP": "Can one support cutting government spending on public services while supporting same-sex marriage?",
    },
    {
        "topic": "redistribution_womensrole",
        "consistent_EL_SP": "Can you support wealth redistribution while supporting equal roles for women in business and government?",
        "consistent_ER_ST": "Can you support protecting private wealth while believing women's primary role is in the home?",
        "cross_EL_ST": "Can you support wealth redistribution while believing women's primary role is in the home?",
        "cross_ER_SP": "Can you support protecting private wealth while supporting equal roles for women in business and government?",
    },
    {
        "topic": "jobguarantee_adoption",
        "consistent_EL_SP": "Is it coherent to support a government guarantee of jobs for all while supporting same-sex couples' right to adopt?",
        "consistent_ER_ST": "Is it coherent to oppose a government job guarantee while opposing same-sex couples' right to adopt?",
        "cross_EL_ST": "Is it coherent to support a government guarantee of jobs for all while opposing same-sex couples' right to adopt?",
        "cross_ER_SP": "Is it coherent to oppose a government job guarantee while supporting same-sex couples' right to adopt?",
    },
    {
        "topic": "taxation_transrights",
        "consistent_EL_SP": "Is it consistent to support higher taxes on the wealthy while supporting legal protections for transgender people?",
        "consistent_ER_ST": "Is it consistent to support lower taxes on the wealthy while opposing legal recognition of transgender identity?",
        "cross_EL_ST": "Is it consistent to support higher taxes on the wealthy while opposing legal recognition of transgender identity?",
        "cross_ER_SP": "Is it consistent to support lower taxes on the wealthy while supporting legal protections for transgender people?",
    },
    {
        "topic": "unions_deathpenalty",
        "consistent_EL_SP": "Should unions be strengthened while the death penalty is abolished?",
        "consistent_ER_ST": "Should union power be reduced while the death penalty is retained?",
        "cross_EL_ST": "Should unions be strengthened while the death penalty is retained?",
        "cross_ER_SP": "Should union power be reduced while the death penalty is abolished?",
    },
    {
        "topic": "insurance_schoolprayer",
        "consistent_EL_SP": "Is it possible to support a government medical insurance plan while opposing organized prayer in public schools?",
        "consistent_ER_ST": "Is it possible to oppose a government medical insurance plan while supporting organized prayer in public schools?",
        "cross_EL_ST": "Is it possible to support a government medical insurance plan while supporting organized prayer in public schools?",
        "cross_ER_SP": "Is it possible to oppose a government medical insurance plan while opposing organized prayer in public schools?",
    },
    {
        "topic": "welfare_sexedu",
        "consistent_EL_SP": "Can you support expanding welfare programs while supporting comprehensive sex education in public schools?",
        "consistent_ER_ST": "Can you support cutting welfare programs while supporting abstinence-only sex education in public schools?",
        "cross_EL_ST": "Can you support expanding welfare programs while supporting abstinence-only sex education in public schools?",
        "cross_ER_SP": "Can you support cutting welfare programs while supporting comprehensive sex education in public schools?",
    },
    {
        "topic": "minwage_contraception",
        "consistent_EL_SP": "Can one support raising the minimum wage while supporting government funding for contraception access?",
        "consistent_ER_ST": "Can one support a market-set wage while opposing government funding for contraception access?",
        "cross_EL_ST": "Can one support raising the minimum wage while opposing government funding for contraception access?",
        "cross_ER_SP": "Can one support a market-set wage while supporting government funding for contraception access?",
    },
    {
        "topic": "childcare_religiousexemption",
        "consistent_EL_SP": "Can you support publicly funded childcare while opposing religious exemptions from anti-discrimination laws?",
        "consistent_ER_ST": "Can you oppose publicly funded childcare while supporting religious exemptions from anti-discrimination laws?",
        "cross_EL_ST": "Can you support publicly funded childcare while supporting religious exemptions from anti-discrimination laws?",
        "cross_ER_SP": "Can you oppose publicly funded childcare while opposing religious exemptions from anti-discrimination laws?",
    },
    {
        "topic": "pension_genderpay",
        "consistent_EL_SP": "Is it coherent to support an expanded public pension system while supporting mandatory gender pay equity laws?",
        "consistent_ER_ST": "Is it coherent to support privatizing pensions while opposing mandatory gender pay equity laws?",
        "cross_EL_ST": "Is it coherent to support an expanded public pension system while opposing mandatory gender pay equity laws?",
        "cross_ER_SP": "Is it coherent to support privatizing pensions while supporting mandatory gender pay equity laws?",
    },
    {
        "topic": "envreg_deathpenalty",
        "consistent_EL_SP": "Can one support strict environmental regulation while opposing the death penalty?",
        "consistent_ER_ST": "Can one support deregulation while supporting the death penalty?",
        "cross_EL_ST": "Can one support strict environmental regulation while supporting the death penalty?",
        "cross_ER_SP": "Can one support deregulation while opposing the death penalty?",
    },
]

MATCHED_CELL_MAP = {
    "consistent_EL_SP": "coupled",
    "consistent_ER_ST": "coupled",
    "cross_EL_ST": "cross",
    "cross_ER_SP": "cross",
}

matched_prompts = []
for t in MATCHED_QUESTION_SETS:
    for cell, family in MATCHED_CELL_MAP.items():
        matched_prompts.append({
            "topic": t["topic"],
            "cell": cell,
            "family": family,
            "question": t[cell],
        })

assert len(matched_prompts) == 48
print("\nMatched design:", len(MATCHED_QUESTION_SETS), "topics × 4 cells =", len(matched_prompts), "prompts")

HEDGING_WORDS = [
    "however", "although", "though", "nevertheless", "nonetheless",
    "on the other hand", "that said", "depends", "it depends",
    "complex", "nuanced", "complicated", "multifaceted",
    "balance", "balanced", "both sides", "multiple perspectives",
    "arguably", "debatable", "controversial",
    "some argue", "others argue", "some believe", "others believe",
    "pros and cons", "trade-off", "tradeoff",
    "not necessarily", "it's important to consider", "important to note",
    "whereas",
]
_HEDGE_SORTED = sorted(HEDGING_WORDS, key=len, reverse=True)

def hedging_density(text):
    t, cnt = text.lower(), 0
    for w in _HEDGE_SORTED:
        pat = r"\b" + re.escape(w) + r"\b"
        cnt += len(re.findall(pat, t))
        t = re.sub(pat, " ", t)
    n = len(text.split())
    return cnt / n * 100 if n else 0.0

RUN_HEDGING_JUDGE = bool(os.environ.get("OPENAI_API_KEY"))
JUDGE_MODEL = "gpt-4.1-mini-2025-04-14"
MAX_CONCURRENT_JUDGES = 50

HEDGING_JUDGE_PROMPT = (
    "You are evaluating how much a text hedges — how much it avoids committing "
    "to a clear position by qualifying, balancing opposing views, expressing "
    "ambivalence, or deferring judgment.\n\n"
    "Question:\n{question}\n\nAnswer:\n{answer}\n\n"
    "Rate hedging from 0 to 100, where 0 is fully committed and 100 is maximally "
    "hedged. Respond with a single number from 0 to 100."
)

class OptionalOpenAIJudge:
    def __init__(self, model_name, prompt_template):
        from openai import AsyncOpenAI
        self.client = AsyncOpenAI()
        self.model = model_name
        self.prompt_template = prompt_template

    async def __call__(self, question, answer):
        c = await self.client.chat.completions.create(
            model=self.model,
            messages=[{
                "role": "user",
                "content": self.prompt_template.format(question=question, answer=answer),
            }],
            max_tokens=1,
            temperature=0,
            logprobs=True,
            top_logprobs=20,
            seed=0,
        )
        try:
            lp = c.choices[0].logprobs.content[0].top_logprobs
        except (IndexError, AttributeError):
            return None
        probs = {el.token: float(math.exp(el.logprob)) for el in lp}
        s = total = 0.0
        for k, v in probs.items():
            try:
                ik = int(k)
            except ValueError:
                continue
            if 0 <= ik <= 100:
                s += ik * v
                total += v
        return None if total < 0.25 else s / total

async def _judge_many(judge, questions, answers):
    sem = asyncio.Semaphore(MAX_CONCURRENT_JUDGES)
    async def _one(q, a):
        async with sem:
            return await judge(q, a)
    return await asyncio.gather(*[_one(q, a) for q, a in zip(questions, answers)])

print("\n" + "=" * 110)
print("CENTER ANALYSIS 2 — MATCHED 2x2")
print("=" * 110)

cuda_cleanup("before matched 2x2 generation")

matched_jobs = []
for prompt_idx, row in enumerate(matched_prompts):
    for rollout in range(N_ROLLOUTS):
        matched_jobs.append({
            "prompt_idx": prompt_idx,
            "rollout": rollout,
            **row,
        })

matched_rows = []
MATCHED_BATCH = choose_batch_size(target=64, minimum=8)
pos = 0
pbar = tqdm(total=len(matched_jobs), desc="matched 2x2 generation", unit="seq")
while pos < len(matched_jobs):
    take = min(MATCHED_BATCH, len(matched_jobs) - pos)
    batch = matched_jobs[pos:pos+take]
    questions = [x["question"] for x in batch]
    keys = [f"matched|{x['prompt_idx']}|{x['rollout']}" for x in batch]
    try:
        gout = unified_generate_batch(
            questions,
            job_keys=keys,
            alpha_e=None,
            alpha_s=None,
            decode=True,
        )
        for j, h, L, ans in zip(
            batch,
            gout["entropy"],
            gout["lengths"],
            gout["answers"],
        ):
            matched_rows.append({
                **j,
                "answer": ans,
                "n_tokens": int(L),
                "n_words": len(ans.split()),
                "mean_entropy": float(h),
                "hedging_lexicon": hedging_density(ans),
                "hedging_judge": np.nan,
                "truncated": bool(int(L) >= MAX_NEW_TOKENS),
                "entropy_definition": ENTROPY_DEFINITION,
                "max_new_tokens": MAX_NEW_TOKENS,
                "temperature": GEN_TEMPERATURE,
                "top_p": GEN_TOP_P,
            })
        del gout
        pos += take
        pbar.update(take)
    except torch.cuda.OutOfMemoryError:
        cuda_cleanup("matched OOM")
        if MATCHED_BATCH <= MIN_BATCH_SIZE:
            raise
        MATCHED_BATCH = max(MIN_BATCH_SIZE, MATCHED_BATCH // 2)
        print("Matched retry BATCH_SIZE =", MATCHED_BATCH)
pbar.close()

matched_raw = pd.DataFrame(matched_rows)

if RUN_HEDGING_JUDGE and len(matched_raw):
    print("Running optional hedging judge...")
    judge = OptionalOpenAIJudge(JUDGE_MODEL, HEDGING_JUDGE_PROMPT)
    try:
        import nest_asyncio
        nest_asyncio.apply()
    except Exception:
        pass
    vals = asyncio.get_event_loop().run_until_complete(
        _judge_many(
            judge,
            matched_raw["question"].tolist(),
            matched_raw["answer"].tolist(),
        )
    )
    matched_raw["hedging_judge"] = vals
else:
    print("OPENAI_API_KEY not available -> hedging_judge skipped; entropy + lexical hedging still run.")

matched_raw.to_csv(MATCHED_DIR / "matched_2x2_raw.csv", index=False)
print("Matched truncation =", f"{matched_raw['truncated'].mean():.2%}")

cuda_cleanup("after matched 2x2 generation")

# ============================================================
# 42. MATCHED 2x2 — ORIGINAL STATISTICS + ITEM/TOPIC DRIVERS
# ============================================================

matched_unit = (
    matched_raw.groupby(["topic","cell","family"], as_index=False)
    .agg(
        mean_entropy=("mean_entropy","mean"),
        hedging_lexicon=("hedging_lexicon","mean"),
        hedging_judge=("hedging_judge","mean"),
        n_tokens=("n_tokens","mean"),
        truncation_rate=("truncated","mean"),
        n_rollouts=("rollout","count"),
    )
)

metric_cols = ["mean_entropy","hedging_lexicon"]
if matched_unit["hedging_judge"].notna().any():
    metric_cols.append("hedging_judge")

print("\nMatched four-cell distributions:")
display(
    matched_unit.groupby("cell")[metric_cols]
    .agg(["mean","std"])
    .round(4)
)

matched_test_rows = []
ECON_PAIRS = [
    ("EL", "cross_EL_ST", "consistent_EL_SP"),
    ("ER", "cross_ER_SP", "consistent_ER_ST"),
]
SOCI_PAIRS = [
    ("SP", "cross_ER_SP", "consistent_EL_SP"),
    ("ST", "cross_EL_ST", "consistent_ER_ST"),
]

for comparison_type, pairs in [
    ("econ_fixed", ECON_PAIRS),
    ("social_fixed", SOCI_PAIRS),
]:
    for axis, cross_cell, coupled_cell in pairs:
        c1 = matched_unit[matched_unit["cell"] == cross_cell].set_index("topic")
        c0 = matched_unit[matched_unit["cell"] == coupled_cell].set_index("topic")
        common = c1.index.intersection(c0.index)
        for metric in metric_cols:
            d = c1.loc[common, metric] - c0.loc[common, metric]
            stat, p = _safe_wilcoxon(d, alternative="greater")
            matched_test_rows.append({
                "comparison_type": comparison_type,
                "fixed_axis_value": axis,
                "cross_cell": cross_cell,
                "coupled_cell": coupled_cell,
                "metric": metric,
                "mean_delta_cross_minus_coupled": float(d.mean()),
                "median_delta": float(d.median()),
                "positive_share": float((d > 0).mean()),
                "W": stat,
                "p_one_sided": p,
                "n_topics": len(common),
            })

matched_tests = pd.DataFrame(matched_test_rows)
matched_tests.to_csv(MATCHED_DIR / "matched_simple_effect_tests.csv", index=False)

print("\nMatched simple effects:")
display(matched_tests.round(5))

topic_2x2_rows = []
for topic, d in matched_unit.groupby("topic"):
    vals = d.set_index("cell")
    if not set(MATCHED_CELL_MAP).issubset(vals.index):
        continue
    row = {"topic": topic}
    for metric in metric_cols:
        elsp = float(vals.loc["consistent_EL_SP", metric])
        erst = float(vals.loc["consistent_ER_ST", metric])
        elst = float(vals.loc["cross_EL_ST", metric])
        ersp = float(vals.loc["cross_ER_SP", metric])
        cross_minus_coupled = ((elst + ersp) - (elsp + erst)) / 2
        factorial_interaction = ((elsp + erst) - (elst + ersp)) / 4
        row[f"{metric}__EL_cross"] = elst - elsp
        row[f"{metric}__ER_cross"] = ersp - erst
        row[f"{metric}__cross_minus_coupled"] = cross_minus_coupled
        row[f"{metric}__factorial_interaction"] = factorial_interaction
    topic_2x2_rows.append(row)

matched_topic_effects = pd.DataFrame(topic_2x2_rows)
matched_topic_effects.to_csv(MATCHED_DIR / "matched_topic_item_effects.csv", index=False)

factorial_summary_rows = []
for metric in metric_cols:
    x = matched_topic_effects[f"{metric}__cross_minus_coupled"]
    stat, p = _safe_wilcoxon(x, alternative="greater")
    factorial_summary_rows.append({
        "metric": metric,
        "mean_cross_minus_coupled": float(x.mean()),
        "median_cross_minus_coupled": float(x.median()),
        "positive_share": float((x > 0).mean()),
        "W": stat,
        "p_one_sided": p,
        "n_topics": len(x),
    })

matched_factorial_summary = pd.DataFrame(factorial_summary_rows)
matched_factorial_summary.to_csv(
    MATCHED_DIR / "matched_factorial_cross_vs_coupled_summary.csv",
    index=False,
)

print("\nMatched cross-vs-coupled summary across 12 topics:")
display(matched_factorial_summary.round(5))

ranked_matched = matched_topic_effects.sort_values(
    "mean_entropy__cross_minus_coupled",
    ascending=True,
)
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(
    ranked_matched["topic"],
    ranked_matched["mean_entropy__cross_minus_coupled"],
)
ax.axvline(0, linewidth=1)
ax.set_xlabel("Cross-pressured minus coupled entropy")
ax.set_title("Matched 2×2: Which Topic Pairs Drive the Entropy Difference?")
plt.tight_layout()
plt.savefig(MATCHED_DIR / "matched_topic_driver_rank.png", dpi=180, bbox_inches="tight")
plt.show()
plt.close(fig)

cuda_cleanup("after matched statistics", reset_peak=False)

# ============================================================
# 43. UNIFIED FINAL DASHBOARD
# ============================================================

print("\n" + "=" * 120)
print("UNIFIED FINAL DASHBOARD")
print("=" * 120)

print("\n[A] ORIGINAL MAIN RESULTS — PRESERVED")
print("Raw vector cosine:", f"{RAW_COSINE:+.6f}")
print("Orthogonal cosine:", f"{ORTH_COSINE:+.8f}")
print("\nStrong simple effects:")
display(strong_results[strong_results["metric"] == "mean_entropy"].round(5))
print("\nFixed-axis sweeps:")
display(fixed_results.round(5))
print("\nNon-additive cross vs consistent:")
display(nonadd_results.round(5))
print("\n2D interaction:")
display(interaction_table.round(6))
print("\nTopic heterogeneity:")
display(topic_interaction.round(5))

print("\n[B] NEW ITEM-LEVEL DRIVER ANALYSIS")
display(
    item_anchor[
        ["question_idx","topic","wvs_source","ER_ST_minus_baseline",
         "EL_cross_effect","ER_cross_effect","cross_minus_coupled"]
    ].head(15).round(5)
)

print("\n[C1] PROMPT-STATE COMPOSITION CONTROL")
display(
    corner_summary[
        np.isclose(corner_summary["alpha_mag"], ANCHOR)
    ].round(4)
)
print("\nPrompt-state entropy linkage:")
display(mechanism_reg.round(6))

print("\n[C2] PRIMARY GENERATION-TIME COMPOSITION/COLLAPSE")
display(
    trajectory_corner_summary[
        np.isclose(trajectory_corner_summary["alpha_mag"], ANCHOR)
    ].round(4)
)
print("\nGeneration-time AXIS-CONDITIONED mechanism tests (EL/ER + SP/ST):")
display(trajectory_side_tests.round(5))
print("\nGeneration-time FOUR-CORNER entropy linkage (Spearman + OLS):")
display(trajectory_corner_linkage.round(5))
print("\nGeneration-time AXIS-CONDITIONED entropy linkage (Spearman + OLS):")
display(trajectory_side_linkage.round(5))

print("\n[D] MATCHED 2×2 — 48 PROMPTS")
display(matched_factorial_summary.round(5))

print("\nOutputs:")
print(" MAIN:", RUN_DIR)
print(" EXTENSIONS:", EXT_DIR)
print(" TRAJECTORY:", TRAJ_DIR)
print(" MATCHED:", MATCHED_DIR)

# ============================================================
# 44. FINAL CUDA CLEANUP
# ============================================================

print("\nFinal GPU cleanup...")

for _name in ["model", "tokenizer", "v_e_gpu", "v_s_gpu"]:
    if _name in globals():
        try:
            del globals()[_name]
        except Exception:
            pass

cuda_cleanup("final", reset_peak=False, ipc=True)

print("\nALL DONE.")
```
